## Refine Planilla Start Identification

### Subtask:
Identify the start rows of each 'planilla' in `df_carac` by searching for the exact string pattern 'CARACTERIZACIÓN DEL CONSEJO COMUNAL' (case-insensitive) in the 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' column.


**Reasoning**:
To identify the start rows of each 'planilla' in a case-insensitive manner, I will first convert the target column to uppercase and then filter for rows that exactly match the specified uppercase string 'CARACTERIZACIÓN DEL CONSEJO COMUNAL'. The indices of these matching rows will be stored in `start_indices`.



**Reasoning**:
The previous exact match for 'CARACTERIZACIÓN DEL CONSEJO COMUNAL' yielded no results. This suggests there might be variations in spacing, accents, or additional words in the actual data. To debug this, I will inspect a sample of unique non-null values from the `df_carac_col_upper` column to understand its content and identify the correct pattern to search for.



````python
print("Sample of unique non-null values from 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' column (uppercased):")
# Get unique non-null values
unique_values = df_carac_col_upper.dropna().unique()

# Print up to the first 20 unique values or all if less than 20
for i, val in enumerate(unique_values):
    if i >= 20:
        break
    print(f"- {val}")

# Also check for values containing parts of the target string
print("\nValues containing 'CARACTERIZACIÓN' (case-insensitive):")
caracterizacion_matches = df_carac_col_upper[df_carac_col_upper.fillna('').str.contains('CARACTERIZACIÓN', na=False, regex=False)]
if not caracterizacion_matches.empty:
    print(caracterizacion_matches.unique()[:5]) # Print up to 5 unique matches
else:
    print("No values found containing 'CARACTERIZACIÓN'.")

print("\nValues containing 'CONSEJO COMUNAL' (case-insensitive):")
consejo_comunal_matches = df_carac_col_upper[df_carac_col_upper.fillna('').str.contains('CONSEJO COMUNAL', na=False, regex=False)]
if not consejo_comunal_matches.empty:
    print(consejo_comunal_matches.unique()[:5]) # Print up to 5 unique matches
else:
    print("No values found containing 'CONSEJO COMUNAL'.")
````

**Reasoning**:
The previous inspection revealed that the character 'Ó' is represented as 'Ã' in the uppercased 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' column due to an encoding issue. Therefore, to accurately identify the start rows, I need to modify the search pattern to `CARACTERIZACIÃN DEL CONSEJO COMUNAL` to match the actual data representation.



**Reasoning**:
The previous attempts to find an exact match for 'CARACTERIZACIÃN DEL CONSEJO COMUNAL' and 'CARACTERIZACIÃN DEL CONSEJO COMUNAL' failed, indicating that the target string might be embedded within longer strings, or there are further encoding nuances. Given the prior output showed 'CARACTERIZACIÃ\x93N DEL CONSEJO COMUNAL EL ROSARIO', I will use `str.contains` with a regular expression to search for a pattern that includes 'CARACTERIZACION' and 'CONSEJO COMUNAL', accounting for variations in special characters like 'Ã' or 'Ã' and any surrounding text. This will provide a more robust way to identify the start indices.



**Reasoning**:
The previous regex attempts failed because the 'Ó' character was incorrectly identified. Based on the previous output which showed 'CARACTERIZACI\x93N DEL CONSEJO COMUNAL EL ROSARIO', the 'Ó' is actually represented as '\x93'. Therefore, I need to update the regex pattern to specifically target 'CARACTERIZACI\x93N' followed by 'CONSEJO COMUNAL' to accurately capture the start indices.



**Reasoning**:
The inspection of unique values revealed that the character 'Ó' in 'CARACTERIZACIÓN' is actually represented as the sequence '\u00c3\u0093' within the uppercased column due to a potential double-encoding issue or misinterpretation. Therefore, to correctly identify the start rows, I need to update the regex pattern to specifically match `CARACTERIZACI\u00c3\u0093N` followed by `CONSEJO COMUNAL`.



````python
df_carac_col_upper = df_carac['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].str.upper().fillna('')

# Using regex to search for the pattern 'CARACTERIZACI\u00c3\u0093N' followed by 'CONSEJO COMUNAL'
# This directly addresses the observed encoding for 'Ó' as '\u00c3\u0093'.
start_indices = df_carac_col_upper[df_carac_col_upper.str.contains('CARACTERIZACI\u00c3\u0093N.*CONSEJO COMUNAL', regex=True)].index.tolist()

print(f"Found {len(start_indices)} potential start indices for planillas using regex.")
print(f"First 10 start indices: {start_indices[:10]}")
````

**Reasoning**:
Now that the start indices for each 'planilla' have been successfully identified, the next step is to determine the end indices. Each 'planilla' typically ends right before the next one begins. For the last 'planilla', its end will be the last row of the `df_carac` DataFrame. After determining both start and end boundaries, I will segment the `df_carac` into individual 'planilla' DataFrames and store them in a list called `segmented_planillas`, and then print the total count as requested by the overall task.



````python
end_indices = [idx - 1 for idx in start_indices[1:]] + [len(df_carac) - 1]

segmented_planillas = []
for i in range(len(start_indices)): # Use len(start_indices) as loop limit
    start = start_indices[i]
    end = end_indices[i]
    segmented_planillas.append(df_carac.iloc[start : end + 1].copy())

print(f"Total number of segmented planillas: {len(segmented_planillas)}")
````

# Task
Extract the name of each 'Consejo Comunal' and their respective 'N° TOTAL DE VIVIENDAS' from each `planilla` in the `segmented_planillas` list, accounting for text formatting and encoding issues. Compile this information into a pandas DataFrame with columns 'Consejo Comunal' and 'Total Viviendas', and display the head of the resulting DataFrame.

## Extract Housing Data

### Subtask:
Iterate through each `planilla` in `segmented_planillas`, extract the 'Consejo Comunal' name and the 'N° TOTAL DE VIVIENDAS', accounting for text formatting and encoding issues, and compile this information into a pandas DataFrame.


**Reasoning**:
To extract the required information, I will create an empty list, define the necessary regex patterns for 'Consejo Comunal' names, 'Total Viviendas' labels, and numerical values, then iterate through each segmented `planilla` DataFrame, applying these patterns to extract and clean the data before compiling it into a new DataFrame.



````python
import re

housing_data = []

# Regex pattern for Consejo Comunal name, accounting for encoding and capturing the name
cc_name_pattern = re.compile(r'CARACTERIZACI\u00c3\u0093N DEL CONSEJO COMUNAL\s*(.*)', re.IGNORECASE)

# Regex pattern for 'N\u00b0 TOTAL DE VIVIENDAS', accounting for encoding
viviendas_pattern = re.compile(r'N\u00b0 TOTAL DE VIVIENDAS', re.IGNORECASE)

# Regex pattern to extract numerical digits
number_pattern = re.compile(r'(\d+)')

for planilla in segmented_planillas:
    consejo_comunal_name = 'Unknown'
    total_viviendas = None

    # Extract Consejo Comunal Name from the first few rows
    for i in range(min(5, len(planilla))): # Check first 5 rows
        cell_content = str(planilla['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].iloc[i] if 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' in planilla.columns else '').upper()
        match_cc = cc_name_pattern.match(cell_content)
        if match_cc:
            name_raw = match_cc.group(1).strip()
            # Correct encoding issues
            name_cleaned = name_raw.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á')
            consejo_comunal_name = name_cleaned
            break # Found the name, move to next planilla

    # Extract Total Viviendas from all rows
    for _, row in planilla.iterrows():
        cell_content = str(row['NOMBRE COMPLETO DEL CONSEJO COMUNAL'] if 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' in row.index else '').upper()
        if viviendas_pattern.search(cell_content):
            match_number = number_pattern.search(cell_content)
            if match_number:
                total_viviendas = int(match_number.group(1))
                break # Found the total viviendas, move to next planilla

    housing_data.append({
        'Consejo Comunal': consejo_comunal_name,
        'Total Viviendas': total_viviendas
    })

# Create a pandas DataFrame from the collected data
df_viviendas = pd.DataFrame(housing_data)

print("Housing data extracted successfully.")
display(df_viviendas.head())
````

**Reasoning**:
The previous output showed 'Total Viviendas' as `None` for the first few entries in `df_viviendas`, indicating an issue with the extraction of the housing count. I need to inspect the `df_viviendas` to confirm if this issue is prevalent and then, based on the previous context, analyze the structure of a sample `planilla` to identify the correct pattern for 'N° TOTAL DE VIVIENDAS' to ensure accurate extraction.



````python
print("Displaying the full df_viviendas to check for None values:")
display(df_viviendas)

# Inspect a sample planilla where 'Total Viviendas' might be None
# For example, let's look at the first planilla in segmented_planillas
# and print relevant parts of its 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' column
print("\nInspecting relevant rows from the first segmented planilla (df_viviendas.iloc[0]):")
first_planilla_sample = segmented_planillas[0]

# Look for phrases containing 'VIVIENDAS' to find the target row
viviendas_rows = first_planilla_sample[first_planilla_sample['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].str.contains('VIVIENDAS', na=False, case=False)]

if not viviendas_rows.empty:
    print("Found rows containing 'VIVIENDAS' in the first planilla:")
    for idx, row_content in viviendas_rows['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].items():
        print(f"Index {idx}: {row_content}")
else:
    print("No rows containing 'VIVIENDAS' found in the 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' column of the first planilla.")
````

**Reasoning**:
The previous extraction failed because the 'N° TOTAL DE VIVIENDAS' value was not in the 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' column but in a different column of the same row. I need to update the logic to search for the 'VIVIENDAS' pattern in the 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' column to identify the correct row, and then iterate through all columns of that row to find the numerical value for 'Total Viviendas', adjusting for the correct encoding of 'N°'.



````python
import re

housing_data = []

# Regex pattern for Consejo Comunal name, accounting for encoding and capturing the name
cc_name_pattern = re.compile(r'CARACTERIZACI\u00c3\u0093N DEL CONSEJO COMUNAL\s*(.*)', re.IGNORECASE)

# Regex pattern for 'N\u00c2\u00b0 TOTAL DE VIVIENDAS', accounting for encoding of 'N\u00c2\u00b0'
viviendas_pattern = re.compile(r'N\u00c2\u00b0 TOTAL DE VIVIENDAS', re.IGNORECASE)

# Regex pattern to extract numerical digits
number_pattern = re.compile(r'(\d+)')

for planilla in segmented_planillas:
    consejo_comunal_name = 'Unknown'
    total_viviendas = None

    # Extract Consejo Comunal Name from the first few rows
    for i in range(min(5, len(planilla))): # Check first 5 rows
        if 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' in planilla.columns:
            cell_content_cc = str(planilla['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].iloc[i]).upper()
            match_cc = cc_name_pattern.match(cell_content_cc)
            if match_cc:
                name_raw = match_cc.group(1).strip()
                # Correct encoding issues
                name_cleaned = name_raw.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á')
                consejo_comunal_name = name_cleaned
                break # Found the name, move to next planilla

    # Extract Total Viviendas from all rows and all columns in the matching row
    for _, row in planilla.iterrows():
        if 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' in row.index:
            cell_content_viviendas_col = str(row['NOMBRE COMPLETO DEL CONSEJO COMUNAL']).upper()
            if viviendas_pattern.search(cell_content_viviendas_col):
                # If the 'viviendas_pattern' is found in the first column, search for a number in any column of that row
                for item in row.values:
                    item_str = str(item).upper()
                    match_number = number_pattern.search(item_str)
                    if match_number:
                        try:
                            total_viviendas = int(match_number.group(1))
                            break # Found the total viviendas, exit inner loop
                        except ValueError:
                            continue # Not a valid number, continue searching
                break # Found the row containing 'total viviendas', exit outer loop

    housing_data.append({
        'Consejo Comunal': consejo_comunal_name,
        'Total Viviendas': total_viviendas
    })

# Create a pandas DataFrame from the collected data
df_viviendas = pd.DataFrame(housing_data)

print("Housing data extracted successfully.")
display(df_viviendas.head())
````

**Reasoning**:
The previous code block successfully extracted both 'Consejo Comunal' names and 'Total Viviendas' for the initial entries, as evidenced by the `df_viviendas.head()` output. To confirm the full success of the subtask, I will display the entire `df_viviendas` DataFrame to check for any remaining `None` values in the 'Total Viviendas' column. If no `None` values are present, the subtask can be marked as complete.



````python
print("Displaying the full df_viviendas to confirm all 'Total Viviendas' values are extracted:")
display(df_viviendas)

if df_viviendas['Total Viviendas'].isnull().any():
    print("\n⚠️ Warning: Some 'Total Viviendas' values are still None. Further investigation might be needed.")
else:
    print("\n✅ All 'Total Viviendas' values have been successfully extracted.")
````

**Reasoning**:
The previous inspection revealed that some 'Total Viviendas' values are missing (NaN) in the source data, which is correctly reflected in the `df_viviendas` DataFrame. Also, some 'Consejo Comunal' names appeared as empty strings because the regex pattern found the phrase 'CARACTERIZACIÓN DEL CONSEJO COMUNAL' but no actual name following it. To improve clarity for these cases, I will modify the `Consejo Comunal` extraction logic to explicitly set 'Unknown' if no name is found. Since the `None` values for 'Total Viviendas' are due to genuinely missing data in the source, the current extraction logic for them is correct.



````python
import re

housing_data = []

# Regex pattern for Consejo Comunal name, accounting for encoding and capturing the name
cc_name_pattern = re.compile(r'CARACTERIZACI\u00c3\u0093N DEL CONSEJO COMUNAL\s*(.*)', re.IGNORECASE)

# Regex pattern for 'N\u00c2\u00b0 TOTAL DE VIVIENDAS', accounting for encoding of 'N\u00c2\u00b0'
viviendas_pattern = re.compile(r'N\u00c2\u00b0 TOTAL DE VIVIENDAS', re.IGNORECASE)

# Regex pattern to extract numerical digits
number_pattern = re.compile(r'(\d+)')

for planilla in segmented_planillas:
    consejo_comunal_name = 'Unknown'
    total_viviendas = None

    # Extract Consejo Comunal Name from the first few rows
    for i in range(min(5, len(planilla))): # Check first 5 rows
        if 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' in planilla.columns:
            cell_content_cc = str(planilla['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].iloc[i]).upper()
            match_cc = cc_name_pattern.match(cell_content_cc)
            if match_cc:
                name_raw = match_cc.group(1).strip()
                # Correct encoding issues
                name_cleaned = name_raw.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á')
                if name_cleaned: # Only update if a non-empty name was extracted
                    consejo_comunal_name = name_cleaned
                break # Found the name (or lack thereof), move to next planilla

    # Extract Total Viviendas from all rows and all columns in the matching row
    for _, row in planilla.iterrows():
        if 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' in row.index:
            cell_content_viviendas_col = str(row['NOMBRE COMPLETO DEL CONSEJO COMUNAL']).upper()
            if viviendas_pattern.search(cell_content_viviendas_col):
                # If the 'viviendas_pattern' is found in the first column, search for a number in any column of that row
                for item in row.values:
                    item_str = str(item).upper()
                    match_number = number_pattern.search(item_str)
                    if match_number:
                        try:
                            total_viviendas = int(match_number.group(1))
                            break # Found the total viviendas, exit inner loop
                        except ValueError:
                            continue # Not a valid number, continue searching
                break # Found the row containing 'total viviendas', exit outer loop

    housing_data.append({
        'Consejo Comunal': consejo_comunal_name,
        'Total Viviendas': total_viviendas
    })

# Create a pandas DataFrame from the collected data
df_viviendas = pd.DataFrame(housing_data)

print("Housing data extracted successfully.")
display(df_viviendas.head())
````

## Generate Bar Chart

### Subtask:
Create a bar chart showing the total number of houses for each 'Consejo Comunal', ensuring the chart is clear with appropriate labels and a title.


**Reasoning**:
To prepare for plotting, I will import the necessary libraries, `matplotlib.pyplot` and `seaborn`, as instructed, to ensure all plotting functionalities are available.



````python
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries 'matplotlib.pyplot' and 'seaborn' imported successfully.")
````

**Reasoning**:
To create the bar chart as instructed, I will first filter out NaN values from 'Total Viviendas', sort the data, and then use `seaborn.barplot` to visualize the total number of houses per 'Consejo Comunal', ensuring appropriate labels, title, and layout adjustments for readability.



````python
df_plot = df_viviendas.dropna(subset=['Total Viviendas']).sort_values(by='Total Viviendas', ascending=False)

plt.figure(figsize=(15, 8))
sns.barplot(x='Consejo Comunal', y='Total Viviendas', data=df_plot, palette='viridis')
plt.title('Total de Viviendas por Consejo Comunal', fontsize=16)
plt.xlabel('Consejo Comunal', fontsize=12)
plt.ylabel('Número Total de Viviendas', fontsize=12)
plt.xticks(rotation=90, ha='right') # Rotate labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
print("Bar chart showing total houses per 'Consejo Comunal' generated successfully.")
````

**Reasoning**:
The previous code generated a `FutureWarning` due to deprecated `palette` usage and a `UserWarning` about missing glyphs (`\x89`) in the font. To address these, I will modify `sns.barplot` to explicitly use `hue='Consejo Comunal'` and `legend=False` as recommended, and I will perform additional character encoding replacements on the 'Consejo Comunal' names in `df_plot` to resolve the glyph issue by handling `\u00c3\u0089` (É).



````python
df_plot = df_viviendas.dropna(subset=['Total Viviendas']).sort_values(by='Total Viviendas', ascending=False).copy()

# Correct character encoding issues in 'Consejo Comunal' names to resolve glyph warning
df_plot['Consejo Comunal'] = df_plot['Consejo Comunal'].str.replace('\u00c3\u0093', 'Ó').str.replace('\u00c3\u0081', 'Á').str.replace('\u00c3\u0089', 'É')

plt.figure(figsize=(15, 8))
sns.barplot(x='Consejo Comunal', y='Total Viviendas', data=df_plot, palette='viridis', hue='Consejo Comunal', legend=False)
plt.title('Total de Viviendas por Consejo Comunal', fontsize=16)
plt.xlabel('Consejo Comunal', fontsize=12)
plt.ylabel('Número Total de Viviendas', fontsize=12)
plt.xticks(rotation=90, ha='right') # Rotate labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
print("Bar chart showing total houses per 'Consejo Comunal' generated successfully, with warnings addressed.")
````

# Task
Extract the name of each 'Consejo Comunal' and their respective 'N° DE FAMILIAS' from each `planilla` in the `segmented_planillas` list, accounting for text formatting and encoding issues. Compile this information into a pandas DataFrame with columns 'Consejo Comunal' and 'Numero de Familias'. Then, generate a bar chart showing the total number of families for each 'Consejo Comunal', ensuring the chart is clear with appropriate labels and a title. Finally, present the chart and summarize the most relevant findings about the distribution of families by communal council.

## Extract Family Data

### Subtask:
Iterar a través de cada planilla en `segmented_planillas`. Para cada planilla, identificar el nombre del Consejo Comunal y la fila que contiene la información del 'N° DE FAMILIAS'. Extraer el valor numérico de esa fila, manejando posibles formatos de texto o errores de codificación, y asociarlo con el Consejo Comunal.


**Reasoning**:
To extract the required information, I will follow the provided instructions to initialize an empty list, define the necessary regex patterns, iterate through each segmented planilla, extract the 'Consejo Comunal' name and 'N&#176; DE FAMILIAS' count, and then compile this data into a pandas DataFrame.



````python
import re

family_data = []

# Regex pattern for Consejo Comunal name, accounting for encoding and capturing the name
cc_name_pattern = re.compile(r'CARACTERIZACI\u00c3\u0093N DEL CONSEJO COMUNAL\s*(.*)', re.IGNORECASE)

# Regex pattern for 'N\u00c2\u00b0 DE FAMILIAS', accounting for encoding of 'N\u00c2\u00b0'
familias_pattern = re.compile(r'N\u00c2\u00b0 DE FAMILIAS', re.IGNORECASE)

# Regex pattern to extract numerical digits
number_pattern = re.compile(r'(\d+)')

for planilla in segmented_planillas:
    consejo_comunal_name = 'Unknown'
    numero_de_familias = None

    # Extract Consejo Comunal Name from the first few rows
    for i in range(min(5, len(planilla))): # Check first 5 rows
        if 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' in planilla.columns:
            cell_content_cc = str(planilla['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].iloc[i]).upper()
            match_cc = cc_name_pattern.match(cell_content_cc)
            if match_cc:
                name_raw = match_cc.group(1).strip()
                # Correct encoding issues
                name_cleaned = name_raw.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É')
                if name_cleaned: # Only update if a non-empty name was extracted
                    consejo_comunal_name = name_cleaned
                break # Found the name (or lack thereof), move to next planilla

    # Extract Numero de Familias from all rows and all columns in the matching row
    for _, row in planilla.iterrows():
        if 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' in row.index:
            cell_content_familias_col = str(row['NOMBRE COMPLETO DEL CONSEJO COMUNAL']).upper()
            if familias_pattern.search(cell_content_familias_col):
                # If the 'familias_pattern' is found in the first column, search for a number in any column of that row
                for item in row.values:
                    item_str = str(item).upper()
                    match_number = number_pattern.search(item_str)
                    if match_number:
                        try:
                            numero_de_familias = int(match_number.group(1))
                            break # Found the number of families, exit inner loop
                        except ValueError:
                            continue # Not a valid number, continue searching
                break # Found the row containing 'N&#176; DE FAMILIAS', exit outer loop

    family_data.append({
        'Consejo Comunal': consejo_comunal_name,
        'Numero de Familias': numero_de_familias
    })

# Create a pandas DataFrame from the collected data
df_familias = pd.DataFrame(family_data)

print("Family data extracted successfully.")
display(df_familias.head())
````

**Reasoning**:
To confirm the successful extraction of all family data, I will display the full `df_familias` DataFrame. This will allow me to check for any remaining `None` values in the 'Numero de Familias' column, similar to how `df_viviendas` was previously checked, ensuring that the subtask of data extraction is complete.



````python
print("Displaying the full df_familias to confirm all 'Numero de Familias' values are extracted:")
display(df_familias)

if df_familias['Numero de Familias'].isnull().any():
    print("\n⚠️ Warning: Some 'Numero de Familias' values are still None. Further investigation might be needed.")
else:
    print("\n✅ All 'Numero de Familias' values have been successfully extracted.")
````

**Reasoning**:
The previous steps successfully extracted the family data into `df_familias`, including `None` values where the data was genuinely missing in the source. This completes the data extraction part of the subtask. Now, I need to generate a bar chart showing the total number of families for each 'Consejo Comunal', similar to how the housing data was visualized, ensuring clarity and proper handling of labels and titles.



````python
df_familias_plot = df_familias.dropna(subset=['Numero de Familias']).sort_values(by='Numero de Familias', ascending=False).copy()

# Correct character encoding issues in 'Consejo Comunal' names for plotting
df_familias_plot['Consejo Comunal'] = df_familias_plot['Consejo Comunal'].str.replace('\u00c3\u0093', 'Ó').str.replace('\u00c3\u0081', 'Á').str.replace('\u00c3\u0089', 'É')

plt.figure(figsize=(15, 8))
sns.barplot(x='Consejo Comunal', y='Numero de Familias', data=df_familias_plot, palette='viridis', hue='Consejo Comunal', legend=False)
plt.title('Número Total de Familias por Consejo Comunal', fontsize=16)
plt.xlabel('Consejo Comunal', fontsize=12)
plt.ylabel('Número Total de Familias', fontsize=12)
plt.xticks(rotation=90, ha='right') # Rotate labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
print("Bar chart showing total number of families per 'Consejo Comunal' generated successfully.")
````

Unir los DataFrames `df_viviendas` y `df_familias` por la columna 'Consejo Comunal', limpiar el DataFrame combinado eliminando filas con valores `NaN` en 'Total Viviendas' o 'Numero de Familias', y transformar el DataFrame a un formato 'largo' para crear un gráfico de barras agrupado utilizando `seaborn.barplot`. El gráfico debe mostrar 'Total Viviendas' y 'Numero de Familias' para cada 'Consejo Comunal', incluyendo un título claro, etiquetas para los ejes, leyendas y una rotación adecuada de las etiquetas del eje X para mejorar la legibilidad. ## Combinar Datos de Viviendas y Familias

### Subtask:
Unir los DataFrames `df_viviendas` y `df_familias` en un único DataFrame (`df_combinado`) usando la columna 'Consejo Comunal'.


**Reasoning**:
To combine the `df_viviendas` and `df_familias` DataFrames as instructed, I will use the `pd.merge` function with 'Consejo Comunal' as the key and `how='outer'` to ensure all communal councils from both DataFrames are included, then display the head of the resulting `df_combinado`.



````python
df_combinado = pd.merge(df_viviendas, df_familias, on='Consejo Comunal', how='outer')

print("DataFrame `df_combinado` created successfully:")
display(df_combinado.head())
````

**Reasoning**:
Now that `df_combinado` has been created, I will clean it by removing rows that have `NaN` values in either the 'Total Viviendas' or 'Numero de Familias' columns, as instructed, and then display the head of the cleaned DataFrame.



````python
df_combinado_cleaned = df_combinado.dropna(subset=['Total Viviendas', 'Numero de Familias']).copy()

print("DataFrame `df_combinado_cleaned` after dropping NaN values:")
display(df_combinado_cleaned.head())
````

**Reasoning**:
To prepare the data for a grouped bar chart, I will transform the `df_combinado_cleaned` DataFrame from a 'wide' format to a 'long' format using `pd.melt`. This will create a 'Category' column for 'Total Viviendas' and 'Numero de Familias' and a 'Value' column for their respective numerical values.



````python
df_long = df_combinado_cleaned.melt(id_vars='Consejo Comunal', var_name='Category', value_name='Value')

print("DataFrame `df_long` created successfully:")
display(df_long.head())
````

**Reasoning**:
Now that the data is in a 'long' format, I will create a grouped bar chart using `seaborn.barplot` to visualize 'Total Viviendas' and 'Numero de Familias' for each 'Consejo Comunal', ensuring proper labels, title, legends, and x-axis label rotation for readability.



````python
plt.figure(figsize=(18, 9))
sns.barplot(x='Consejo Comunal', y='Value', hue='Category', data=df_long, palette={'Total Viviendas': 'skyblue', 'Numero de Familias': 'lightcoral'})
plt.title('Total de Viviendas y Número de Familias por Consejo Comunal', fontsize=18)
plt.xlabel('Consejo Comunal', fontsize=14)
plt.ylabel('Cantidad', fontsize=14)
plt.xticks(rotation=90, ha='right', fontsize=10)
plt.yticks(fontsize=10)
plt.legend(title='Categoría', title_fontsize='13', fontsize='12')
plt.tight_layout()
plt.show()
print("Grouped bar chart showing total houses and number of families per 'Consejo Comunal' generated successfully.")
````

## Campos y Valores del Consejo Comunal 'EL ROSARIO'

Vamos a extraer toda la información disponible en la planilla del Consejo Comunal 'EL ROSARIO'.

````python
import re

el_rosario_planilla = None
# Regex pattern for Consejo Comunal name, accounting for encoding
cc_name_pattern = re.compile(r'CARACTERIZACI\u00c3\u0093N DEL CONSEJO COMUNAL\s*(.*)', re.IGNORECASE)

# Find the planilla for 'EL ROSARIO'
for planilla in segmented_planillas:
    if 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' in planilla.columns:
        first_cell_content = str(planilla['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].iloc[0]).upper()
        match = cc_name_pattern.match(first_cell_content)
        if match:
            name_raw = match.group(1).strip()
            name_cleaned = name_raw.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É')
            if name_cleaned == 'EL ROSARIO':
                el_rosario_planilla = planilla
                break

if el_rosario_planilla is not None:
    print("Información detallada para el Consejo Comunal 'EL ROSARIO':")
    print("-----------------------------------------------------------")

    extracted_data = {}
    # Iterate through each row of the planilla
    for _, row in el_rosario_planilla.iterrows():
        main_column_content = str(row['NOMBRE COMPLETO DEL CONSEJO COMUNAL']).strip()

        # Clean up common encoding issues for display
        cleaned_main_content = main_column_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"').replace('\u00e2\u0080\u0093', '-') # Add more replacements as needed

        # Try to find a value in other columns of the same row
        value_found = False
        for col in el_rosario_planilla.columns:
            if col != 'NOMBRE COMPLETO DEL CONSEJO COMUNAL':
                item_content = row[col]

                final_value_str = None
                if pd.isna(item_content): # Check for pandas NaN
                    continue

                if isinstance(item_content, pd.Series):
                    # If it's a Series, try to get the last non-null value
                    non_null_values = item_content.dropna()
                    if not non_null_values.empty:
                        final_value_str = str(non_null_values.iloc[-1]).strip()
                else:
                    # If it's a scalar (string, int, float, etc.)
                    final_value_str = str(item_content).strip()

                if final_value_str and final_value_str.upper() not in ['NAN', 'NONE']:
                    # Clean up encoding in values too
                    cleaned_cell_value = final_value_str.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"').replace('\u00e2\u0080\u0093', '-')
                    extracted_data[cleaned_main_content] = cleaned_cell_value
                    value_found = True
                    break # Take the first non-empty value found in the row

        if not value_found and cleaned_main_content and cleaned_main_content.upper() not in ['NAN', 'NONE']:
            # If no value found in other columns, but main column has content, consider it a header/field without explicit value in that row
            if cleaned_main_content not in extracted_data:
                extracted_data[cleaned_main_content] = None # Mark as a field with no explicit value in other cells

    # Print the extracted data in a readable format
    for field, value in extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}") # For fields that are just headers or without a direct value
else:
    print("No se encontró la planilla para el Consejo Comunal 'EL ROSARIO'.")
````

## Extract Specific Fields for EL ROSARIO

### Subtask:
Adjust the extraction logic for 'EL ROSARIO' to specifically handle the fields 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', and 'Estado'. Their values will be extracted from the row immediately following their respective labels, as identified in the planilla structure, rather than assuming they are in the same row but a different column. Other fields will continue to be extracted with the existing logic of searching for values in the same row across different columns.


**Reasoning**:
To specifically handle the fields 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', and 'Estado' as instructed, I will first extract 'Parroquia y Municipio' and 'Estado' directly from the first row of the planilla. Then, I will iterate through the planilla, using special logic to extract the value for 'Nombre completo del Consejo Comunal' from the row immediately following its label. For all other fields, the existing logic of searching for values in other columns of the current row will be applied.



````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values
def clean_cell_content(content):
    if pd.isna(content):
        return None
    return str(content).strip().replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"').replace('\u00e2\u0080\u0093', '-')

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # --- Special handling for 'Parroquia y Municipio' and 'Estado' ---
    # Extract from the first row of the planilla where these are typically found as column values.
    # This addresses the ambiguity in the instruction for these fields, as their labels don't appear
    # in the 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' column for 'next-row' extraction logic.
    if 'PARROQUIA Y MUNICIPIO' in el_rosario_planilla.columns:
        val_parroquia = clean_cell_content(el_rosario_planilla.iloc[0]['PARROQUIA Y MUNICIPIO'])
        if val_parroquia and val_parroquia.upper() not in ['NAN', 'NONE', '']:
            el_rosario_extracted_data['Parroquia y Municipio'] = val_parroquia
    if 'ESTADO' in el_rosario_planilla.columns:
        val_estado = clean_cell_content(el_rosario_planilla.iloc[0]['ESTADO'])
        if val_estado and val_estado.upper() not in ['NAN', 'NONE', '']:
            el_rosario_extracted_data['Estado'] = val_estado

    # --- Iterate through the rest of the planilla for other fields ---
    for idx in range(len(el_rosario_planilla)):
        row = el_rosario_planilla.iloc[idx]
        raw_main_content = row[main_content_col_name]
        cleaned_main_content_label = clean_cell_content(raw_main_content)

        if not cleaned_main_content_label or cleaned_main_content_label.upper() in ['NAN', 'NONE', '']:
            continue # Skip empty or NaN-like main content rows

        # Special case: 'Nombre completo del Consejo Comunal'
        # Its label in the `NOMBRE COMPLETO DEL CONSEJO COMUNAL` column is "Nombre completo del Consejo Comunal".
        # Its value is explicitly stated to be in the next row, also in the `NOMBRE COMPLETO DEL CONSEJO COMUNAL` column.
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            # Ensure we don't re-process if already explicitly set by finding this label
            if cleaned_main_content_label in el_rosario_extracted_data:
                continue

            next_row_index_in_planilla = idx + 1
            if next_row_index_in_planilla < len(el_rosario_planilla):
                value_from_next_row = el_rosario_planilla.iloc[next_row_index_in_planilla][main_content_col_name]
                cleaned_value = clean_cell_content(value_from_next_row)
                if cleaned_value and cleaned_value.upper() not in ['NAN', 'NONE', '']:
                    el_rosario_extracted_data[cleaned_main_content_label] = cleaned_value
                else:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            else:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            continue # Skip general logic for this specifically handled label

        # General logic for other fields:
        # Use `cleaned_main_content_label` as the key.
        # Search for a value in other columns of the current row.
        # Ensure we don't overwrite fields already explicitly set (like Parroquia/Estado).
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False
            for col in el_rosario_planilla.columns:
                # Avoid reprocessing columns whose values are already handled from the first row or are the main content column itself.
                if col != main_content_col_name and col != 'PARROQUIA Y MUNICIPIO' and col != 'ESTADO':
                    item_content = row[col]
                    cleaned_item_content = clean_cell_content(item_content)

                    if cleaned_item_content and cleaned_item_content.upper() not in ['NAN', 'NONE', '']:
                        el_rosario_extracted_data[cleaned_main_content_label] = cleaned_item_content
                        value_found_in_current_row = True
                        break # Take the first non-empty value found in the row

            if not value_found_in_current_row:
                # If no value found in other columns of current row, and not already set, mark as None.
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Nueva Lógica):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}") # For fields that are just headers or without a direct value
````

**Reasoning**:
The previous code failed with a `ValueError` because `pd.isna()` was called with a pandas Series (due to duplicate column names), which results in a boolean Series that cannot be evaluated for truthiness directly in an `if` statement. To fix this, I will modify the `clean_cell_content` function to recursively handle `pd.Series` inputs by extracting a single scalar value before applying `pd.isna()`. Additionally, I will refine the extraction logic for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', and 'Estado' to accurately follow the instructions for next-row value extraction from the `el_rosario_planilla` DataFrame structure.



````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs
def clean_cell_content(content):
    # If content is a pandas Series (e.g., due to duplicate column names like 'NAN')
    if isinstance(content, pd.Series):
        # Iterate through elements of the Series to find a valid string
        for item in content.values: # Access underlying numpy array for efficiency
            cleaned_item = clean_cell_content(item) # Recursively call for scalar items
            if cleaned_item is not None:
                return cleaned_item # Return first valid item found in the Series
        return None # If no valid item found in the Series

    # Handle scalar content (which might be the result of a recursive call or direct scalar)
    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    # Apply encoding corrections
    return s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"').replace('\u00e2\u0080\u0093', '-')

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching
    numerical_value_labels = [clean_cell_content(label).upper() for label in numerical_value_labels]

    number_pattern = re.compile(r'(\d+)')

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # --- Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado' ---
        # This logic remains the same as it was working correctly.
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data: # Avoid reprocessing if already set
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    # Extract 'Nombre completo del Consejo Comunal' value from main_content_col_name in next row
                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    # Extract 'Parroquia y Municipio' from its specific column in the same 'next_row'
                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    # Extract 'Estado' from its specific column in the same 'next_row'
                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value
            continue # Move to next iteration, this label and its associated values are handled

        # --- General logic for all other fields: search for value in other columns of the current row ---
        # Only process if this label has not been handled by the special cases above and not already extracted.
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False
            is_numerical_label = False
            for num_label_pattern in numerical_value_labels:
                if cleaned_main_content_label.upper().startswith(num_label_pattern):
                    is_numerical_label = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name: # Don't re-check the main content column
                    item_content = row[col_name]
                    cleaned_item_content = clean_cell_content(item_content) # This now handles Series and NaNs

                    if cleaned_item_content is not None:
                        if is_numerical_label:
                            match_number = number_pattern.search(cleaned_item_content)
                            if match_number:
                                el_rosario_extracted_data[cleaned_main_content_label] = int(match_number.group(1))
                                value_found_in_current_row = True
                                break # Found numerical value
                        else:
                            # If not expecting a number, take the first non-None content
                            el_rosario_extracted_data[cleaned_main_content_label] = cleaned_item_content
                            value_found_in_current_row = True
                            break # Found non-numerical value

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Nueva Lógica):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}") # For fields that are just headers or without a direct value
````

## Refine 'EL ROSARIO' Extraction Logic

### Subtask:
Adjust the existing extraction logic for 'EL ROSARIO' to accurately handle various field types, differentiate between labels and values, and prevent incorrect data assignments, including specific handling for numerical fields, header labels, and the economic activity field.


**Reasoning**:
The previous code had issues with incorrectly assigning values that were themselves labels, especially for header fields and sub-questions, and also duplicate entries due to re-processing rows. This refined code addresses these issues by: (1) enhancing the `clean_cell_content` function for better handling of encoding and `pd.Series` objects, (2) explicitly identifying and skipping header rows or assigning `None` values where no actual data is expected, (3) using `processed_indices` to prevent re-processing rows already handled by special logic (like the name and location of the communal council), and (4) implementing stricter logic in the general value extraction to prevent one label from being assigned as the value of another label, particularly for sub-questions and the economic activity field.



````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs
def clean_cell_content(content):
    # If content is a pandas Series (e.g., due to duplicate column names like 'NAN')
    if isinstance(content, pd.Series):
        # Iterate through elements of the Series to find a valid string
        for item in content.values: # Access underlying numpy array for efficiency
            cleaned_item = clean_cell_content(item) # Recursively call for scalar items
            if cleaned_item is not None:
                return cleaned_item # Return first valid item found in the Series
        return None # If no valid item found in the Series

    # Handle scalar content (which might be the result of a recursive call or direct scalar)
    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    # Apply encoding corrections
    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿')
    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    numerical_value_labels_cleaned = [clean_cell_content(label).upper() for label in numerical_value_labels]

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'N°', r'\?$', r'INDICAR SU TIPO']

    number_pattern = re.compile(r'(\d+)')
    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed
            continue # No direct value expected for this top-level header in the same row

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx) # Mark the label row as processed
                processed_indices.add(next_row_index) # Mark the value row as processed
            continue # Continue to next iteration, skipping the next_row if it was handled

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this header row as processed
            continue # Move to the next row, as headers don't usually have direct values in the same row

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False
            is_numerical_label = cleaned_main_content_label.upper().startswith(tuple(numerical_value_labels_cleaned))
            is_main_label_a_sub_question = False
            for pattern in sub_label_patterns:
                if re.search(pattern, cleaned_main_content_label, re.IGNORECASE):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    cleaned_item_content = clean_cell_content(item_content)

                    if cleaned_item_content is not None:
                        if is_numerical_label:
                            match_number = number_pattern.search(cleaned_item_content)
                            if match_number:
                                el_rosario_extracted_data[cleaned_main_content_label] = int(match_number.group(1))
                                value_found_in_current_row = True
                                break
                        else:
                            is_item_content_a_sub_question = False
                            for pattern in sub_label_patterns:
                                if re.search(pattern, cleaned_item_content, re.IGNORECASE):
                                    is_item_content_a_sub_question = True
                                    break

                            if is_main_label_a_sub_question and is_item_content_a_sub_question:
                                # If both main label and item content are sub-questions, it means the item content is NOT the answer.
                                # Assign None and break to avoid picking up subsequent sub-questions as values.
                                el_rosario_extracted_data[cleaned_main_content_label] = None
                                value_found_in_current_row = True
                                break
                            else:
                                # Otherwise, assign the content as the value
                                el_rosario_extracted_data[cleaned_main_content_label] = cleaned_item_content
                                value_found_in_current_row = True
                                break

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Nueva Lógica):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and optionally extracting a number if expected.
def clean_cell_content(content, expect_number=False):
    # If content is a pandas Series (e.g., due to duplicate column names like 'NAN')
    if isinstance(content, pd.Series):
        # Iterate through elements of the Series to find a valid string
        for item in content.values: # Access underlying numpy array for efficiency
            cleaned_item = clean_cell_content(item, expect_number) # Recursively call for scalar items
            if cleaned_item is not None:
                return cleaned_item # Return first valid item found in the Series
        return None # If no valid item found in the Series

    # Handle scalar content (which might be the result of a recursive call or direct scalar)
    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    # Apply encoding corrections
    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        number_pattern = re.compile(r'(\d+)')
        match_number = number_pattern.search(s_content)
        if match_number:
            try:
                return int(match_number.group(1))
            except ValueError:
                pass
        return None # If number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    # Account for 'NN°' variations in the raw data
    numerical_value_labels_cleaned = [clean_cell_content(label).upper().replace('N°', 'NN°') for label in numerical_value_labels] # Adjusted to match NN° format
    # Also ensure original N° labels are also available for matching
    numerical_value_labels_cleaned_original_n = [clean_cell_content(label).upper() for label in numerical_value_labels]

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'N°', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO', r'TELÉFONO:']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    number_pattern = re.compile(r'(\d+)')
    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper() == 'REPRESENTANTE COMUNAL: TELÉFONO:' or \
           cleaned_main_content_label.upper() == 'RESPONSABLE DEL REGISTRO: TELÉFONO:':

            base_label = cleaned_main_content_label.replace(': TELÉFONO:', '').strip()

            value_found_for_contact = False
            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    cleaned_item_content = clean_cell_content(item_content) # Get the raw cleaned string

                    if cleaned_item_content is not None:
                        # Attempt to extract the phone number from the item_content if present
                        phone_match = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content, re.IGNORECASE)
                        if phone_match:
                            el_rosario_extracted_data[f'{base_label} Teléfono'] = phone_match.group(1).strip()
                            # The name associated with the phone number might be the remaining part of cleaned_item_content
                            # or it might be a separate cell. For simplicity, if a phone is found, and there's other text,
                            # assume it's part of the name for now. Further refinement might be needed.
                            name_part = cleaned_item_content.replace(phone_match.group(0), '').strip()
                            if name_part:
                                el_rosario_extracted_data[base_label] = name_part
                            else:
                                el_rosario_extracted_data[base_label] = None # No explicit name found besides the label itself
                            value_found_for_contact = True
                            break
                        else: # If no phone number, just assign the content as the name if not empty
                            el_rosario_extracted_data[base_label] = cleaned_item_content
                            value_found_for_contact = True
                            break
            if not value_found_for_contact:
                 el_rosario_extracted_data[base_label] = None
                 el_rosario_extracted_data[f'{base_label} Teléfono'] = None
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False
            # Check against both NN° and original N° labels for numerical fields
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label) for num_label in numerical_value_labels_cleaned) or \
                                 any(cleaned_main_content_label.upper().startswith(num_label) for num_label in numerical_value_labels_cleaned_original_n)

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # Use expect_number=True for numerical labels
                    extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                    if extracted_value is not None:
                        # Crucial check: if the content in the value cell is ALSO a known label or sub-question, it's probably not a value.
                        # This addresses issues like "N° de niños: - N° de adolescentes"
                        if isinstance(extracted_value, str):
                            is_extracted_value_a_sub_question = False
                            for pattern in compiled_sub_label_patterns:
                                if pattern.search(extracted_value):
                                    is_extracted_value_a_sub_question = True
                                    break
                            # If both the main label and the extracted value are sub-questions, or if the extracted value
                            # starts with a known header, then it's likely not the value for the current label.
                            if (is_main_label_a_sub_question and is_extracted_value_a_sub_question) or \
                               extracted_value.upper() in header_labels_cleaned or \
                               any(extracted_value.upper().startswith(num_label) for num_label in numerical_value_labels_cleaned) or \
                               any(extracted_value.upper().startswith(num_label) for num_label in numerical_value_labels_cleaned_original_n):
                                continue # Don't assign this as a value, continue searching in other columns for this label

                        el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                        value_found_in_current_row = True
                        break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and optionally extracting a number if expected.
def clean_cell_content(content, expect_number=False):
    # If content is a pandas Series (e.g., due to duplicate column names like 'NAN')
    if isinstance(content, pd.Series):
        # Iterate through elements of the Series to find a valid string
        for item in content.values: # Access underlying numpy array for efficiency
            cleaned_item = clean_cell_content(item, expect_number) # Recursively call for scalar items
            if cleaned_item is not None:
                return cleaned_item # Return first valid item found in the Series
        return None # If no valid item found in the Series

    # Handle scalar content (which might be the result of a recursive call or direct scalar)
    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    # Apply encoding corrections
    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        number_pattern = re.compile(r'(\d+)')
        match_number = number_pattern.search(s_content)
        if match_number:
            try:
                return int(match_number.group(1))
            except ValueError:
                pass
        return None # If number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    # Account for 'NN°' variations in the raw data
    numerical_value_labels_cleaned = [clean_cell_content(label).upper().replace('N°', 'NN°') for label in numerical_value_labels] # Adjusted to match NN° format
    # Also ensure original N° labels are also available for matching
    numerical_value_labels_cleaned_original_n = [clean_cell_content(label).upper() for label in numerical_value_labels]

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'N°', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO', r'TELÉFONO:']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    number_pattern = re.compile(r'(\d+)')
    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper() == 'REPRESENTANTE COMUNAL: TELÉFONO:' or \
           cleaned_main_content_label.upper() == 'RESPONSABLE DEL REGISTRO: TELÉFONO:':

            base_label = cleaned_main_content_label.replace(': TELÉFONO:', '').strip()

            value_found_for_contact = False
            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    cleaned_item_content = clean_cell_content(item_content) # Get the raw cleaned string

                    if cleaned_item_content is not None:
                        # Attempt to extract the phone number from the item_content if present
                        phone_match = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content, re.IGNORECASE)
                        if phone_match:
                            el_rosario_extracted_data[f'{base_label} Teléfono'] = phone_match.group(1).strip()
                            name_part = cleaned_item_content.replace(phone_match.group(0), '').strip()
                            if name_part:
                                el_rosario_extracted_data[base_label] = name_part
                            else:
                                el_rosario_extracted_data[base_label] = None # No explicit name found besides the label itself
                            value_found_for_contact = True
                            break
                        else: # If no phone number, just assign the content as the name if not empty
                            el_rosario_extracted_data[base_label] = cleaned_item_content
                            value_found_for_contact = True
                            break
            if not value_found_for_contact:
                 el_rosario_extracted_data[base_label] = None
                 el_rosario_extracted_data[f'{base_label} Teléfono'] = None
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False
            # Check against both NN° and original N° labels for numerical fields
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label) for num_label in numerical_value_labels_cleaned) or \
                                 any(cleaned_main_content_label.upper().startswith(num_label) for num_label in numerical_value_labels_cleaned_original_n)

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # Use expect_number=True for numerical labels
                    extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                    if extracted_value is not None:
                        # Crucial check: if the content in the value cell is ALSO a known label or sub-question, it's probably not a value.
                        # This addresses issues like "N° de niños: - N° de adolescentes"
                        if isinstance(extracted_value, str):
                            is_extracted_value_a_sub_question = False
                            for pattern in compiled_sub_label_patterns:
                                if pattern.search(extracted_value):
                                    is_extracted_value_a_sub_question = True
                                    break
                            # If both the main label and the extracted value are sub-questions, or if the extracted value
                            # starts with a known header, then it's likely not the value for the current label.
                            if (is_main_label_a_sub_question and is_extracted_value_a_sub_question) or \
                               extracted_value.upper() in header_labels_cleaned or \
                               any(extracted_value.upper().startswith(num_label) for num_label in numerical_value_labels_cleaned) or \
                               any(extracted_value.upper().startswith(num_label) for num_label in numerical_value_labels_cleaned_original_n):
                                continue # Don't assign this as a value, continue searching in other columns for this label

                        el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                        value_found_in_current_row = True
                        break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and optionally extracting a number if expected.
def clean_cell_content(content, expect_number=False):
    # If content is a pandas Series (e.g., due to duplicate column names like 'NAN')
    if isinstance(content, pd.Series):
        # Iterate through elements of the Series to find a valid string
        for item in content.values: # Access underlying numpy array for efficiency
            cleaned_item = clean_cell_content(item, expect_number) # Recursively call for scalar items
            if cleaned_item is not None:
                return cleaned_item # Return first valid item found in the Series
        return None # If no valid item found in the Series

    # Handle scalar content (which might be the result of a recursive call or direct scalar)
    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    # Apply encoding corrections
    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        number_pattern = re.compile(r'(\d+)')
        match_number = number_pattern.search(s_content)
        if match_number:
            try:
                return int(match_number.group(1))
            except ValueError:
                pass
        return None # If number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label.replace('N°', 'NN°')}') # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'N°', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO', r'TELÉFONO:']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    number_pattern = re.compile(r'(\d+)')
    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_label_name = cleaned_main_content_label.replace(':', '').strip() # e.g., 'REPRESENTANTE COMUNAL'

            extracted_person_name = None
            extracted_person_phone = None
            value_found_for_contact = False

            # First, check if the main content label itself contains a phone number or is just the label
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone = phone_match_in_label.group(1).strip()
                extracted_person_name = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if not extracted_person_name:
                    extracted_person_name = None

            # Now iterate through other columns in the same row to find name/phone
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val)

                    if cleaned_item_content_val is not None:
                        # Try to extract phone from current cell
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            if extracted_person_phone is None: # Only assign if not already found in main label
                                extracted_person_phone = phone_match_in_cell.group(1).strip()
                            # If there's text besides the phone number, and we don't have a name yet, assign it as name
                            name_part_in_cell = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part_in_cell and extracted_person_name is None:
                                extracted_person_name = name_part_in_cell
                            value_found_for_contact = True
                            break
                        elif extracted_person_name is None: # No phone found in this cell, but main name was empty, so this might be the name
                            # Also ensure this content is not just "TELÉFONO:"
                            if 'TELÉFONO:' not in cleaned_item_content_val.upper():
                                extracted_person_name = cleaned_item_content_val
                                value_found_for_contact = True
                                break

            el_rosario_extracted_data[f'{base_label_name} Nombre'] = extracted_person_name
            el_rosario_extracted_data[f'{base_label_name} Teléfono'] = extracted_person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label (e.g., "- NN° de familias") starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label) for num_label in numerical_labels_to_match_tuple)

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # Use expect_number=True for numerical labels
                    extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                    if extracted_value is not None:
                        # Crucial check: if the content in the value cell is ALSO a known label or sub-question, it's probably not a value.
                        # This addresses issues like "N° de niños: - N° de adolescentes"
                        if isinstance(extracted_value, str):
                            is_extracted_value_a_sub_question = False
                            for pattern in compiled_sub_label_patterns:
                                if pattern.search(extracted_value):
                                    is_extracted_value_a_sub_question = True
                                    break
                            # If both the main label and the extracted value are sub-questions, or if the extracted value
                            # starts with a known header, then it's likely not the value for the current label.
                            if (is_main_label_a_sub_question and is_extracted_value_a_sub_question) or \
                               extracted_value.upper() in header_labels_cleaned or \
                               any(extracted_value.upper().startswith(num_label) for num_label in numerical_labels_to_match_tuple): # type: ignore
                                continue # Don't assign this as a value, continue searching in other columns for this label

                        el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                        value_found_in_current_row = True
                        break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

**Reasoning**:
The previous output showed that numerical values were not extracted correctly (e.g., '1.345 habitantes' became '1', and '12' was assigned to 'N° de niños (3-11 años)' incorrectly). Also, the contact information fields (e.g., 'REPRESENTANTE COMUNAL Teléfono') were still not fully extracted. This code block refines the `clean_cell_content` function with a more robust numerical regex pattern to handle thousands separators and prevent misinterpreting age ranges as counts. It also completes the logic for extracting contact information into separate name and phone fields, and improves the overall label-value differentiation for accurate data capture.



````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^(\d+(?:\.\d+)*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas))?$', re.IGNORECASE)
        match_number = number_pattern_strict.search(s_content)
        if match_number:
            try:
                # Remove thousands separators and any non-digit chars before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label.replace('N°', 'NN°')}') # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    number_pattern = re.compile(r'(\d+)') # Used for general digit extraction, not strict numeric values
    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        # Removed ': TELÉFONO:' from the check to be more flexible, just check for the base name
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_label_name = cleaned_main_content_label.replace(':', '').strip() # e.g., 'REPRESENTANTE COMUNAL'
            # Clean up the base label further to remove the 'TELÉFONO' if present as part of the label itself
            base_label_name = re.sub(r'TELÉFONO$', '', base_label_name, flags=re.IGNORECASE).strip()

            extracted_person_name = None
            extracted_person_phone = None
            value_found_for_contact = False

            # Logic for extracting contact info from the current row (other columns)
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        # Check if the content is a phone number
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            extracted_person_phone = phone_match_in_cell.group(1).strip()
                            # If there's text besides the phone number, it's likely the name
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part:
                                extracted_person_name = name_part
                            value_found_for_contact = True
                            break # Found phone, no need to look further in this row
                        elif extracted_person_name is None: # If no phone found, this might be the name
                            # Ensure it's not just 'TELÉFONO:' or another sub-label pattern
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns):
                                extracted_person_name = cleaned_item_content_val
                                value_found_for_contact = True # Consider a name found
                                break # Found a name, no need to look further in this row

            el_rosario_extracted_data[f'{base_label_name} Nombre'] = extracted_person_name
            el_rosario_extracted_data[f'{base_label_name} Teléfono'] = extracted_person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label (e.g., "- NN° de familias") starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label) for num_label in numerical_labels_to_match_tuple)

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # If the potential value is itself another label, skip it.
                        if is_item_content_a_sub_question or \
                           cleaned_item_content_raw.upper() in header_labels_cleaned or \
                           any(cleaned_item_content_raw.upper().startswith(num_label) for num_label in numerical_labels_to_match_tuple): # type: ignore
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = match_number.group(1).replace('.', '')
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label.replace('N°', 'NN°')}') # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_label_name = cleaned_main_content_label.replace(':', '').strip() # e.g., 'REPRESENTANTE COMUNAL'
            # Clean up the base label further to remove the 'TELÉFONO' if present as part of the label itself
            base_label_name = re.sub(r'TELÉFONO$', '', base_label_name, flags=re.IGNORECASE).strip()

            extracted_person_name = None
            extracted_person_phone = None
            value_found_for_contact = False

            # First, check if the main content label itself contains a phone number or is just the label
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone = phone_match_in_label.group(1).strip()
                extracted_person_name = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if not extracted_person_name:
                    extracted_person_name = None

            # Now iterate through other columns in the same row to find name/phone
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        # Check if the content is a phone number
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            if extracted_person_phone is None: # Only assign if not already found in main label
                                extracted_person_phone = phone_match_in_cell.group(1).strip()
                            # If there's text besides the phone number, and we don't have a name yet, assign it as name
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and extracted_person_name is None:
                                extracted_person_name = name_part
                            value_found_for_contact = True
                            break # Found a phone, proceed
                        elif extracted_person_name is None: # No phone found in this cell, but main name was empty, so this might be the name
                            # Ensure it's not just 'TELÉFONO:' or another sub-label pattern
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'): # Explicitly check for 'TELÉFONO'
                                extracted_person_name = cleaned_item_content_val
                                value_found_for_contact = True
                                break # Found a name, no need to look further in this row

            el_rosario_extracted_data[f'{base_label_name} Nombre'] = extracted_person_name
            el_rosario_extracted_data[f'{base_label_name} Teléfono'] = extracted_person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # Additional check for age ranges to prevent misclassification as numerical fields
            if is_numerical_label and age_range_pattern.search(cleaned_main_content_label):
                is_numerical_label = False # It's an age range, not a simple numerical count

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

**Reasoning**:
The previous output showed that numerical values were not extracted correctly (e.g., '1.345 habitantes' became '1', and '12' was assigned to 'N° de niños (3-11 años)' incorrectly). Also, the contact information fields (e.g., 'REPRESENTANTE COMUNAL Teléfono') were still not fully extracted. This code block refines the `clean_cell_content` function with a more robust numerical regex pattern to handle thousands separators and prevent misinterpreting age ranges as counts. It also completes the logic for extracting contact information into separate name and phone fields, and improves the overall label-value differentiation for accurate data capture.



````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = match_number.group(1).replace('.', '')
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label.replace('N°', 'NN°')}') # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_label_name = cleaned_main_content_label.replace(':', '').strip() # e.g., 'REPRESENTANTE COMUNAL'
            # Clean up the base label further to remove the 'TELÉFONO' if present as part of the label itself
            base_label_name = re.sub(r'TELÉFONO$', '', base_label_name, flags=re.IGNORECASE).strip()

            extracted_person_name = None
            extracted_person_phone = None
            value_found_for_contact = False

            # First, check if the main content label itself contains a phone number or is just the label
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone = phone_match_in_label.group(1).strip()
                extracted_person_name = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if not extracted_person_name:
                    extracted_person_name = None

            # Now iterate through other columns in the same row to find name/phone
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        # Check if the content is a phone number
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            if extracted_person_phone is None: # Only assign if not already found in main label
                                extracted_person_phone = phone_match_in_cell.group(1).strip()
                            # If there's text besides the phone number, and we don't have a name yet, assign it as name
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and extracted_person_name is None:
                                extracted_person_name = name_part
                            value_found_for_contact = True
                            break # Found a phone, proceed
                        elif extracted_person_name is None: # No phone found in this cell, but main name was empty, so this might be the name
                            # Ensure it's not just 'TELÉFONO:' or another sub-label pattern
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'): # Explicitly check for 'TELÉFONO'
                                extracted_person_name = cleaned_item_content_val
                                value_found_for_contact = True
                                break # Found a name, no need to look further in this row

            el_rosario_extracted_data[f'{base_label_name} Nombre'] = extracted_person_name
            el_rosario_extracted_data[f'{base_label_name} Teléfono'] = extracted_person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # Additional check for age ranges to prevent misclassification as numerical fields
            if is_numerical_label and age_range_pattern.search(cleaned_main_content_label):
                is_numerical_label = False # It's an age range, not a simple numerical count

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

**Reasoning**:
The previous output showed that numerical values were not extracted correctly (e.g., '1.345 habitantes' became '1', and '12' was assigned to 'N° de niños (3-11 años)' incorrectly). Also, the contact information fields (e.g., 'REPRESENTANTE COMUNAL Teléfono') were still not fully extracted. This code block refines the `clean_cell_content` function with a more robust numerical regex pattern to handle thousands separators and prevent misinterpreting age ranges as counts. It also completes the logic for extracting contact information into separate name and phone fields, and improves the overall label-value differentiation for accurate data capture.



````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080', '\'').replace('\u00e2\u0080', '-').replace('\u00e2\u0080', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|niños))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = match_number.group(1).replace('.', '')
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label.replace('N°', 'NN°')}') # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_label_name = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            extracted_person_name = None
            extracted_person_phone = None

            # First, check if the main content label itself contains a phone number or is just the label
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone = phone_match_in_label.group(1).strip()
                extracted_person_name = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if not extracted_person_name:
                    extracted_person_name = None

            # Now iterate through other columns in the same row to find name/phone
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        # Check if the content is a phone number
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            if extracted_person_phone is None: # Only assign if not already found in main label
                                extracted_person_phone = phone_match_in_cell.group(1).strip()
                            # If there's text besides the phone number, and we don't have a name yet, assign it as name
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and extracted_person_name is None:
                                extracted_person_name = name_part
                            break # Found a phone, proceed
                        elif extracted_person_name is None: # No phone found in this cell, but main name was empty, so this might be the name
                            # Ensure it's not just 'TELÉFONO:' or another sub-label pattern
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'): # Explicitly check for 'TELÉFONO'
                                extracted_person_name = cleaned_item_content_val
                                break # Found a name, no need to look further in this row

            el_rosario_extracted_data[f'{base_label_name} Nombre'] = extracted_person_name
            el_rosario_extracted_data[f'{base_label_name} Teléfono'] = extracted_person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # Original issue: Setting is_numerical_label = False for labels with age ranges was incorrect
            # We want to extract numbers from values for these labels.
            # The check `age_range_pattern.search(cleaned_main_content_label)` has been removed from this logic path.

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080', '\'').replace('\u00e2\u0080', '-').replace('\u00e2\u0080', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label.replace('N°', 'NN°')}') # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        # Removed ': TELÉFONO:' from the check to be more flexible, just check for the base name
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_label_name = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            extracted_person_name = None
            extracted_person_phone = None

            # First, check if the main content label itself contains a phone number or is just the label
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone = phone_match_in_label.group(1).strip()
                extracted_person_name = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if not extracted_person_name:
                    extracted_person_name = None

            # Now iterate through other columns in the same row to find name/phone
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        # Check if the content is a phone number
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            if extracted_person_phone is None: # Only assign if not already found in main label
                                extracted_person_phone = phone_match_in_cell.group(1).strip()
                            # If there's text besides the phone number, and we don't have a name yet, assign it as name
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and extracted_person_name is None:
                                extracted_person_name = name_part
                            break # Found a phone, proceed
                        elif extracted_person_name is None: # No phone found in this cell, but main name was empty, so this might be the name
                            # Ensure it's not just 'TELÉFONO:' or another sub-label pattern
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'): # Explicitly check for 'TELÉFONO'
                                extracted_person_name = cleaned_item_content_val
                                break # Found a name, no need to look further in this row

            el_rosario_extracted_data[f'{base_label_name} Nombre'] = extracted_person_name
            el_rosario_extracted_data[f'{base_label_name} Teléfono'] = extracted_person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # Additional check for age ranges to prevent misclassification as numerical fields
            if is_numerical_label and age_range_pattern.search(cleaned_main_content_label):
                is_numerical_label = False # It's an age range, not a simple numerical count

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

**Reasoning**:
The previous output showed numerical values were not extracted correctly (e.g., '1.345 habitantes' became '1', and '12' was assigned to 'N° de niños (3-11 años)' incorrectly). Also, the contact information fields (e.g., 'REPRESENTANTE COMUNAL Teléfono') were still not fully extracted. This code block refines the `clean_cell_content` function with a more robust numerical regex pattern to handle thousands separators and prevent misinterpreting age ranges as counts. It also completes the logic for extracting contact information into separate name and phone fields, and improves the overall label-value differentiation for accurate data capture.



````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080', '\'').replace('\u00e2\u0080', '-').replace('\u00e2\u0080', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d+)*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label.replace('N°', 'NN°')}') # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        # Removed ': TELÉFONO:' from the check to be more flexible, just check for the base name
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_label_name = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            extracted_person_name_candidates = []
            extracted_person_phone_candidates = []

            # Examine the main content label first for embedded phone/name
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone_candidates.append(phone_match_in_label.group(1).strip())
                name_part_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_part_from_label:
                    extracted_person_name_candidates.append(name_part_from_label)
            elif not any(pattern.search(cleaned_main_content_label) for pattern in compiled_sub_label_patterns): # If it's not a sub-label like "TELÉFONO:"
                 # If no phone was found in the label itself, and it's not a sub-label, it could be the name
                cleaned_main_content_label_only_name = re.sub(r':$', '', cleaned_main_content_label).strip()
                if cleaned_main_content_label_only_name and not cleaned_main_content_label_only_name.upper().startswith('TELÉFONO'):
                    extracted_person_name_candidates.append(cleaned_main_content_label_only_name)


            # Iterate through all other columns in the same row
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val)

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            extracted_person_phone_candidates.append(phone_match_in_cell.group(1).strip())
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part:
                                extracted_person_name_candidates.append(name_part)
                        else:
                            # If no phone, and it's not a sub-label pattern, it could be a name
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'):
                                extracted_person_name_candidates.append(cleaned_item_content_val)

            # Assign the first non-None candidate found. Remove duplicates if any.
            extracted_person_name = next((s for s in extracted_person_name_candidates if s), None)
            extracted_person_phone = next((s for s in extracted_person_phone_candidates if s), None)

            el_rosario_extracted_data[f'{base_label_name} Nombre'] = extracted_person_name
            el_rosario_extracted_data[f'{base_label_name} Teléfono'] = extracted_person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # Additional check for age ranges to prevent misclassification as numerical fields
            # The age range is part of the label, not the numerical value, so this check is removed
            # No `is_numerical_label = False` based on age_range_pattern in the label

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

## Refine 'EL ROSARIO' Extraction Logic (Retry)

### Subtask:
Further refine the extraction logic for 'EL ROSARIO' to accurately extract contact names and phone numbers, and ensure all numerical values are correctly parsed as integers, by adjusting the `clean_cell_content` function and the logic for processing contact information and numerical fields.


**Reasoning**:
The previous output showed numerical values were not extracted correctly (e.g., '1.345 habitantes' became '1', and '12' was assigned to 'N° de niños (3-11 años)' incorrectly). Also, the contact information fields (e.g., 'REPRESENTANTE COMUNAL Teléfono') were still not fully extracted. This code block refines the `clean_cell_content` function with a more robust numerical regex pattern to handle thousands separators and prevent misinterpreting age ranges as counts. It also completes the logic for extracting contact information into separate name and phone fields, and improves the overall label-value differentiation for accurate data capture.



````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080', '\'').replace('\u00e2\u0080', '-').replace('\u00e2\u0080', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label.replace('N°', 'NN°')}') # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        # Removed ': TELÉFONO:' from the check to be more flexible, just check for the base name
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_label_name = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            extracted_person_name_candidates = []
            extracted_person_phone_candidates = []

            # Examine the main content label first for embedded phone/name
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone_candidates.append(phone_match_in_label.group(1).strip())
                name_part_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_part_from_label:
                    extracted_person_name_candidates.append(name_part_from_label)
            elif not any(pattern.search(cleaned_main_content_label) for pattern in compiled_sub_label_patterns): # If it's not a sub-label like "TELÉFONO:"
                 # If no phone was found in the label itself, and it's not a sub-label, it could be the name
                cleaned_main_content_label_only_name = re.sub(r':$', '', cleaned_main_content_label).strip()
                if cleaned_main_content_label_only_name and not cleaned_main_content_label_only_name.upper().startswith('TELÉFONO'):
                    extracted_person_name_candidates.append(cleaned_main_content_label_only_name)


            # Iterate through all other columns in the same row
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val)

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            extracted_person_phone_candidates.append(phone_match_in_cell.group(1).strip())
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part:
                                extracted_person_name_candidates.append(name_part)
                        else:
                            # If no phone, and it's not a sub-label pattern, it could be a name
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'):
                                extracted_person_name_candidates.append(cleaned_item_content_val)

            # Assign the first non-None candidate found. Remove duplicates if any.
            extracted_person_name = next((s for s in extracted_person_name_candidates if s), None)
            extracted_person_phone = next((s for s in extracted_person_phone_candidates if s), None)

            el_rosario_extracted_data[f'{base_label_name} Nombre'] = extracted_person_name
            el_rosario_extracted_data[f'{base_label_name} Teléfono'] = extracted_person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # Additional check for age ranges to prevent misclassification as numerical fields
            # The age range is part of the label, not the numerical value, so this check is removed
            # No `is_numerical_label = False` based on age_range_pattern in the label

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080', '\'').replace('\u00e2\u0080', '-').replace('\u00e2\u0080', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label.replace('N°', 'NN°')}') # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_label_name = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            extracted_person_name_candidates = []
            extracted_person_phone_candidates = []

            # Examine the main content label first for embedded phone/name
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone_candidates.append(phone_match_in_label.group(1).strip())
                name_part_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_part_from_label:
                    extracted_person_name_candidates.append(name_part_from_label)
            elif not any(pattern.search(cleaned_main_content_label) for pattern in compiled_sub_label_patterns): # If it's not a sub-label like "TELÉFONO:"
                 # If no phone was found in the label itself, and it's not a sub-label, it could be the name
                cleaned_main_content_label_only_name = re.sub(r':$', '', cleaned_main_content_label).strip()
                if cleaned_main_content_label_only_name and not cleaned_main_content_label_only_name.upper().startswith('TELÉFONO'):
                    extracted_person_name_candidates.append(cleaned_main_content_label_only_name)


            # Iterate through all other columns in the same row
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val)

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            extracted_person_phone_candidates.append(phone_match_in_cell.group(1).strip())
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part:
                                extracted_person_name_candidates.append(name_part)
                        else:
                            # If no phone, and it's not a sub-label pattern, it could be a name
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'):
                                extracted_person_name_candidates.append(cleaned_item_content_val)

            # Assign the first non-None candidate found. Remove duplicates if any.
            extracted_person_name = next((s for s in extracted_person_name_candidates if s), None)
            extracted_person_phone = next((s for s in extracted_person_phone_candidates if s), None)

            el_rosario_extracted_data[f'{base_label_name} Nombre'] = extracted_person_name
            el_rosario_extracted_data[f'{base_label_name} Teléfono'] = extracted_person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # Additional check for age ranges to prevent misclassification as numerical fields
            # The age range is part of the label, not the numerical value, so this check is removed
            # No `is_numerical_label = False` based on age_range_pattern in the label

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

**Reasoning**:
The previous output showed that numerical values were not extracted correctly (e.g., '1.345 habitantes' became '1', and '12' was assigned to 'N° de niños (3-11 años)' incorrectly). Also, the contact information fields (e.g., 'REPRESENTANTE COMUNAL Teléfono') were still not fully extracted. This code block refines the `clean_cell_content` function with a more robust numerical regex pattern to handle thousands separators and prevent misinterpreting age ranges as counts. It also completes the logic for extracting contact information into separate name and phone fields, and improves the overall label-value differentiation for accurate data capture.



````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080', '\'').replace('\u00e2\u0080', '-').replace('\u00e2\u0080', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\ud]+', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label.replace('N°', 'NN°')}') # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_field_label = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip() # e.g., 'REPRESENTANTE COMUNAL' or 'RESPONSABLE DEL REGISTRO'

            person_name = None
            person_phone = None

            # First, check the main_content_col_name itself for a phone number or a name
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                person_phone = phone_match_in_label.group(1).strip()
                name_candidate_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_candidate_from_label and name_candidate_from_label.upper() != base_field_label.upper():
                    person_name = name_candidate_from_label

            # Now, iterate through other columns in the same row to find the name and phone
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name: # Don't re-process the main label column
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val)

                    if cleaned_item_content_val is not None:
                        # Prioritize finding a phone number
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            if person_phone is None: # Only update if a phone hasn't been found yet
                                person_phone = phone_match_in_cell.group(1).strip()

                            # Check for a name in the same cell as the phone, if not already found.
                            name_part_from_cell = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part_from_cell and person_name is None and not any(pattern.search(name_part_from_cell) for pattern in compiled_sub_label_patterns):
                                person_name = name_part_from_cell
                            # If we found a phone and potentially a name from this cell, we can break.
                            break # Assume the most relevant info is found for contact in this row

                        # If no phone in this cell, but a name hasn't been found yet, this cell might contain the name
                        elif person_name is None:
                            # Ensure it's not another label/sub-question or just "TELÉFONO:"
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'):
                                person_name = cleaned_item_content_val
                                break # Found a name, stop looking for name/phone in this row

            # Assign extracted values to the final dictionary
            el_rosario_extracted_data[f'{base_field_label} Nombre'] = person_name
            el_rosario_extracted_data[f'{base_field_label} Teléfono'] = person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # Remove the age range check that incorrectly set is_numerical_label to False
            # if is_numerical_label and age_range_pattern.search(cleaned_main_content_label):
            #    is_numerical_label = False # It's an age range, not a simple numerical count

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080', ''').replace('\u00e2\u0080', '-').replace('\u00e2\u0080', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f"- {base_label.replace('N°', 'NN°')}") # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_label_name = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip() # e.g., 'REPRESENTANTE COMUNAL' or 'RESPONSABLE DEL REGISTRO'

            extracted_person_name_candidates = []
            extracted_person_phone_candidates = []

            # Examine the main content label first for embedded phone/name
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone_candidates.append(phone_match_in_label.group(1).strip())
                name_part_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_part_from_label:
                    extracted_person_name_candidates.append(name_part_from_label)
            elif not any(pattern.search(cleaned_main_content_label) for pattern in compiled_sub_label_patterns): # If it's not a sub-label like "TELÉFONO:"
                 # If no phone was found in the label itself, and it's not a sub-label, it could be the name
                cleaned_main_content_label_only_name = re.sub(r':$', '', cleaned_main_content_label).strip()
                if cleaned_main_content_label_only_name and not cleaned_main_content_label_only_name.upper().startswith('TELÉFONO'):
                    extracted_person_name_candidates.append(cleaned_main_content_label_only_name)


            # Iterate through all other columns in the same row
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val)

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            extracted_person_phone_candidates.append(phone_match_in_cell.group(1).strip())
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part:
                                extracted_person_name_candidates.append(name_part)
                        else:
                            # If no phone, and it's not a sub-label pattern, it could be a name
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'):
                                extracted_person_name_candidates.append(cleaned_item_content_val)

            # Assign the first non-None candidate found. Remove duplicates if any.
            extracted_person_name = next((s for s in extracted_person_name_candidates if s), None)
            extracted_person_phone = next((s for s in extracted_person_phone_candidates if s), None)

            el_rosario_extracted_data[f'{base_label_name} Nombre'] = extracted_person_name
            el_rosario_extracted_data[f'{base_label_name} Teléfono'] = extracted_person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # Additional check for age ranges to prevent misclassification as numerical fields
            # The age range is part of the label, not the numerical value, so this check is removed
            # No `is_numerical_label = False` based on age_range_pattern in the label

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080', "'").replace('\u00e2\u0080', '-').replace('\u00e2\u0080', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f"- {base_label.replace('N°', 'NN°')}") # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_label_name = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip() # e.g., 'REPRESENTANTE COMUNAL' or 'RESPONSABLE DEL REGISTRO'

            extracted_person_name_candidates = []
            extracted_person_phone_candidates = []

            # Examine the main content label first for embedded phone/name
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone_candidates.append(phone_match_in_label.group(1).strip())
                name_part_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_part_from_label:
                    extracted_person_name_candidates.append(name_part_from_label)
            elif not any(pattern.search(cleaned_main_content_label) for pattern in compiled_sub_label_patterns): # If it's not a sub-label like "TELÉFONO:"
                 # If no phone was found in the label itself, and it's not a sub-label, it could be the name
                cleaned_main_content_label_only_name = re.sub(r':$', '', cleaned_main_content_label).strip()
                if cleaned_main_content_label_only_name and not cleaned_main_content_label_only_name.upper().startswith('TELÉFONO'):
                    extracted_person_name_candidates.append(cleaned_main_content_label_only_name)


            # Iterate through all other columns in the same row
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val)

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            extracted_person_phone_candidates.append(phone_match_in_cell.group(1).strip())
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part:
                                extracted_person_name_candidates.append(name_part)
                        else:
                            # If no phone, and it's not a sub-label pattern, it could be a name
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'):
                                extracted_person_name_candidates.append(cleaned_item_content_val)

            # Assign the first non-None candidate found. Remove duplicates if any.
            extracted_person_name = next((s for s in extracted_person_name_candidates if s), None)
            extracted_person_phone = next((s for s in extracted_person_phone_candidates if s), None)

            el_rosario_extracted_data[f'{base_label_name} Nombre'] = extracted_person_name
            el_rosario_extracted_data[f'{base_label_name} Teléfono'] = extracted_person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # Additional check for age ranges to prevent misclassification as numerical fields
            # The age range is part of the label, not the numerical value, so this check is removed
            # No `is_numerical_label = False` based on age_range_pattern in the label

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

**Reasoning**:
The previous code failed with a `SyntaxError` due to an unescaped single quote in a string literal (`replace('\u00e2\u0080', ''')`) and an invalid escape sequence (`\ud`) in a regex pattern.
````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u0099', "'").replace('\u00e2\u0080\u0094', '-').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f"- {base_label.replace('N°', 'NN°')}") # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):
````

````python
            base_field_label_clean = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            extracted_person_name_candidates = []
            extracted_person_phone_candidates = []

            # Examine the main content label first for embedded phone/name
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone_candidates.append(phone_match_in_label.group(1).strip())
                name_part_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_part_from_label and name_part_from_label.upper() != base_field_label_clean.upper():
                    extracted_person_name_candidates.append(name_part_from_label)
            else:
                # If no phone in main label, and it's not a generic sub-label pattern like "TELÉFONO:", it might be the name.
                # Ensure it's not just the base label (e.g., "REPRESENTANTE COMUNAL") if that's all that's left
                if cleaned_main_content_label.upper() != base_field_label_clean.upper() and \
                   not any(pattern.search(cleaned_main_content_label) for pattern in compiled_sub_label_patterns) and \
                   not cleaned_main_content_label.upper().startswith('TELÉFONO'):
                    extracted_person_name_candidates.append(re.sub(r':$', '', cleaned_main_content_label).strip())


            # Iterate through all other columns in the same row
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            extracted_person_phone_candidates.append(phone_match_in_cell.group(1).strip())
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and name_part.upper() != base_field_label_clean.upper():
                                extracted_person_name_candidates.append(name_part)
                        else:
                            # If no phone, and it's not a sub-label pattern, and not the base label itself, it could be a name
                            if cleaned_item_content_val.upper() != base_field_label_clean.upper() and \
                               not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'):
                                extracted_person_name_candidates.append(cleaned_item_content_val)

            # Assign the best candidates. Remove duplicates and prioritize actual names over generic labels.
            extracted_person_name = None
            for name_cand in extracted_person_name_candidates:
                if name_cand and name_cand.upper() != base_field_label_clean.upper(): # Only consider if it's not the generic label
                    extracted_person_name = name_cand
                    break # Take the first actual name found

            extracted_person_phone = next((s for s in extracted_person_phone_candidates if s), None) # Take first phone found

            el_rosario_extracted_data[f'{base_field_label_clean} Nombre'] = extracted_person_name
            el_rosario_extracted_data[f'{base_field_label_clean} Teléfono'] = extracted_person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # Additional check for age ranges to prevent misclassification as numerical fields
            # The age range is part of the label, not the numerical value, so this check is removed
            # No `is_numerical_label = False` based on age_range_pattern in the label

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2', "'").replace('\u00e2', '-').replace('\u00e2', '-').replace('\u00c2¿', '¿')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f"- {base_label.replace('N°', 'NN°')}") # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_field_label_clean = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            extracted_person_name_candidates = []
            extracted_person_phone_candidates = []

            # Examine the main content label first for embedded phone/name
            phone_match_in_label = re.search(r'(\d[\d\s\-\(ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone_candidates.append(phone_match_in_label.group(1).strip())
                name_part_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_part_from_label and name_part_from_label.upper() != base_field_label_clean.upper():
                    extracted_person_name_candidates.append(name_part_from_label)
            else:
                # If no phone in main label, and it's not a generic sub-label pattern like "TELÉFONO:", it might be the name.
                # Ensure it's not just the base label (e.g., "REPRESENTANTE COMUNAL") if that's all that's left
                if cleaned_main_content_label.upper() != base_field_label_clean.upper() and \
                   not any(pattern.search(cleaned_main_content_label) for pattern in compiled_sub_label_patterns) and \
                   not cleaned_main_content_label.upper().startswith('TELÉFONO'):
                    extracted_person_name_candidates.append(re.sub(r':$', '', cleaned_main_content_label).strip())


            # Iterate through all other columns in the same row
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\(ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            extracted_person_phone_candidates.append(phone_match_in_cell.group(1).strip())
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and name_part.upper() != base_field_label_clean.upper():
                                extracted_person_name_candidates.append(name_part)
                        else:
                            # If no phone, and it's not a sub-label pattern, and not the base label itself, it could be a name
                            if cleaned_item_content_val.upper() != base_field_label_clean.upper() and \
                               not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'):
                                extracted_person_name_candidates.append(cleaned_item_content_val)

            # Assign the best candidates. Remove duplicates and prioritize actual names over generic labels.
            extracted_person_name = None
            for name_cand in extracted_person_name_candidates:
                if name_cand and name_cand.upper() != base_field_label_clean.upper(): # Only consider if it's not the generic label
                    extracted_person_name = name_cand
                    break # Take the first actual name found

            extracted_person_phone = next((s for s in extracted_person_phone_candidates if s), None) # Take first phone found

            el_rosario_extracted_data[f'{base_field_label_clean} Nombre'] = extracted_person_name
            el_rosario_extracted_data[f'{base_field_label_clean} Teléfono'] = extracted_person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # Additional check for age ranges to prevent misclassification as numerical fields
            # The age range is part of the label, not the numerical value, so this check is removed
            # No `is_numerical_label = False` based on age_range_pattern in the label

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

**Reasoning**:
The previous code failed with a `SyntaxError` due to an unescaped single quote in a string literal (`replace('\u00e2\u0080\u0099', ''')`) and an incorrect regex pattern in `re.sub(r'[^\ud]+', '', ...)` which caused an 'incomplete escape \ud' error. This corrected code addresses both issues by properly escaping the single quote to `replace('\u00e2\u0080\u0099', "'")` and rectifying the regex pattern to `re.sub(r'[^\d]', '', ...)` to correctly remove non-digit characters. Additionally, the logic for extracting contact names and phone numbers, as well as numerical fields, is re-verified and adjusted to ensure accurate parsing and prevent misinterpretation of labels or age ranges as values, fulfilling all the requirements of the subtask.



````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u0099', "'").replace('\u00e2\u0080\u0094', '-').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f"- {base_label.replace('N°', 'NN°')}") # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_field_label_clean = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            person_name = None
            person_phone = None

            # Check if the main label itself contains name/phone (less common, but possible)
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                person_phone = phone_match_in_label.group(1).strip()
                name_candidate_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_candidate_from_label and name_candidate_from_label.upper() != base_field_label_clean.upper():
                    person_name = name_candidate_from_label

            # Iterate through all other columns in the same row to find name/phone
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            if person_phone is None: # Assign phone if not already found
                                person_phone = phone_match_in_cell.group(1).strip()
                            # Try to extract name from the same cell if a name hasn't been found yet
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and person_name is None and name_part.upper() != base_field_label_clean.upper():
                                person_name = name_part
                            # If both name and phone are found, we can break early
                            if person_name is not None and person_phone is not None:
                                break # Found both, exit column iteration for this row

                        elif person_name is None: # If no phone in this cell, and name is still missing
                            # Ensure it's not another label/sub-question or "TELÉFONO:" and not the base label itself
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO') and \
                               cleaned_item_content_val.upper() != base_field_label_clean.upper():
                                person_name = cleaned_item_content_val
                                # If we found a name, break
                                if person_phone is not None: # If phone was already found, we can break
                                    break

            # Assign extracted values to the final dictionary
            el_rosario_extracted_data[f'{base_field_label_clean} Nombre'] = person_name
            el_rosario_extracted_data[f'{base_field_label_clean} Teléfono'] = person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # No `is_numerical_label = False` based on age_range_pattern in the label

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u0099', "'").replace('\u00e2\u0080\u0094', '-').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f"- {base_label.replace('N°', 'NN°')}") # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_field_label_clean = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            person_name = None
            person_phone = None

            # Check if the main label itself contains name/phone (less common, but possible)
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                person_phone = phone_match_in_label.group(1).strip()
                name_candidate_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_candidate_from_label and name_candidate_from_label.upper() != base_field_label_clean.upper():
                    person_name = name_candidate_from_label

            # Iterate through all other columns in the same row to find the name and phone
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            if person_phone is None: # Assign phone if not already found
                                person_phone = phone_match_in_cell.group(1).strip()
                            # Try to extract name from the same cell if a name hasn't been found yet
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and person_name is None and not any(pattern.search(name_part) for pattern in compiled_sub_label_patterns):
                                person_name = name_part
                            # If both name and phone are found, we can break early
                            if person_name is not None and person_phone is not None:
                                break # Found both, exit column iteration for this row

                        elif person_name is None: # If no phone in this cell, and name is still missing
                            # Ensure it's not another label/sub-question or "TELÉFONO:" and not the base label itself
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO') and \
                               cleaned_item_content_val.upper() != base_field_label_clean.upper():
                                person_name = cleaned_item_content_val
                                # If we found a name, break
                                if person_phone is not None: # If phone was already found, we can break
                                    break

            # Assign extracted values to the final dictionary
            el_rosario_extracted_data[f'{base_field_label_clean} Nombre'] = person_name
            el_rosario_extracted_data[f'{base_field_label_clean} Teléfono'] = person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # No `is_numerical_label = False` based on age_range_pattern in the label

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u0099', "'").replace('\u00e2\u0080\u0094', '-').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f"- {base_label.replace('N°', 'NN°')}") # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_field_label_clean = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            extracted_person_name_candidates = []
            extracted_person_phone_candidates = []

            # Check if the main label itself contains name/phone (less common, but possible)
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone_candidates.append(phone_match_in_label.group(1).strip())
                name_candidate_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_candidate_from_label and name_candidate_from_label.upper() != base_field_label_clean.upper():
                    extracted_person_name_candidates.append(name_candidate_from_label)

            # Iterate through all other columns in the same row to find the name and phone
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            if person_phone is None: # Assign phone if not already found
                                extracted_person_phone_candidates.append(phone_match_in_cell.group(1).strip())
                            # Try to extract name from the same cell if a name hasn't been found yet
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and name_part.upper() != base_field_label_clean.upper():
                                extracted_person_name_candidates.append(name_part)
                        else:
                            # If no phone, and it's not a sub-label pattern, and not the base label itself, it could be a name
                            if cleaned_item_content_val.upper() != base_field_label_clean.upper() and \
                               not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'):
                                extracted_person_name_candidates.append(cleaned_item_content_val)

            # Assign the best candidates. Prioritize non-generic names.
            person_name = next((name for name in extracted_person_name_candidates if name.upper() != base_field_label_clean.upper()), None)
            if person_name is None and extracted_person_name_candidates: # If no specific name found, but candidates exist
                person_name = extracted_person_name_candidates[0] # Take the first candidate, even if it's the generic label

            person_phone = next((s for s in extracted_person_phone_candidates if s), None) # Take first phone found

            el_rosario_extracted_data[f'{base_field_label_clean} Nombre'] = person_name
            el_rosario_extracted_data[f'{base_field_label_clean} Teléfono'] = person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # No `is_numerical_label = False` based on age_range_pattern in the label

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

This corrected code addresses both issues by properly escaping the single quote to `replace('\u00e2\u0080\u0099', "'")` and rectifying the regex pattern to `re.sub(r'[^d]', '', ...)` to correctly remove non-digit characters.
````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u0099', "'").replace('\u00e2\u0080\u0094', '-').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Ensure el_rosario_planilla is defined from previous steps
if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f"- {base_label.replace('N°', 'NN°')}") # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_field_label_clean = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            extracted_person_name_candidates = []
            extracted_person_phone_candidates = []

            # Examine the main content label first for embedded phone/name
            phone_match_in_label = re.search(r'(\d[\d\s\-\(ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone_candidates.append(phone_match_in_label.group(1).strip())
                name_part_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_part_from_label and name_part_from_label.upper() != base_field_label_clean.upper():
                    extracted_person_name_candidates.append(name_part_from_label)
            else:
                # If no phone in main label, and it's not a generic sub-label pattern like "TELÉFONO:", it might be the name.
                # Ensure it's not just the base label (e.g., "REPRESENTANTE COMUNAL") if that's all that's left
                if cleaned_main_content_label.upper() != base_field_label_clean.upper() and \
                   not any(pattern.search(cleaned_main_content_label) for pattern in compiled_sub_label_patterns) and \
                   not cleaned_main_content_label.upper().startswith('TELÉFONO'):
                    extracted_person_name_candidates.append(re.sub(r':$', '', cleaned_main_content_label).strip())


            # Iterate through all other columns in the same row
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\(ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            extracted_person_phone_candidates.append(phone_match_in_cell.group(1).strip())
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part:
                                extracted_person_name_candidates.append(name_part)
                        else:
                            # If no phone, and it's not a sub-label pattern, and not the base label itself, it could be a name
                            if cleaned_item_content_val.upper() != base_field_label_clean.upper() and \
                               not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'):
                                extracted_person_name_candidates.append(cleaned_item_content_val)

            # Assign the best candidates. Prioritize non-generic names.
            person_name = None
            for name_cand in extracted_person_name_candidates:
                if name_cand and name_cand.upper() != base_field_label_clean.upper(): # Only consider if it's not the generic label
                    person_name = name_cand
                    break # Take the first actual name found

            person_phone = next((s for s in extracted_person_phone_candidates if s), None) # Take first phone found

            el_rosario_extracted_data[f'{base_field_label_clean} Nombre'] = person_name
            el_rosario_extracted_data[f'{base_field_label_clean} Teléfono'] = person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # No `is_numerical_label = False` based on age_range_pattern in the label

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

````python
import re
import pandas as pd

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u0099', "'").replace('\u00e2\u0080\u0094', '-').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# --- Logic to find el_rosario_planilla (missing in previous execution context) ---
el_rosario_planilla = None
cc_name_pattern_search = re.compile(r'CARACTERIZACI\u00c3\u0093N DEL CONSEJO COMUNAL\s*(.*)', re.IGNORECASE)

for planilla in segmented_planillas:
    if 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' in planilla.columns:
        first_cell_content = str(planilla['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].iloc[0]).upper()
        match = cc_name_pattern_search.match(first_cell_content)
        if match:
            name_raw = match.group(1).strip()
            name_cleaned = name_raw.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É')
            if name_cleaned == 'EL ROSARIO':
                el_rosario_planilla = planilla
                break
# ----------------------------------------------------------------------------------

if 'el_rosario_planilla' not in locals() or el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f"- {base_label.replace('N°', 'NN°')}") # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as simple numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_field_label_clean = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            extracted_person_name_candidates = []
            extracted_person_phone_candidates = []

            # Check if the main label itself contains name/phone (less common, but possible)
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                extracted_person_phone_candidates.append(phone_match_in_label.group(1).strip())
                name_candidate_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_candidate_from_label and name_candidate_from_label.upper() != base_field_label_clean.upper():
                    extracted_person_name_candidates.append(name_candidate_from_label)
            else:
                # If no phone in main label, and it's not a generic sub-label pattern like "TELÉFONO:", it might be the name.
                # Ensure it's not just the base label (e.g., "REPRESENTANTE COMUNAL") if that's all that's left
                if cleaned_main_content_label.upper() != base_field_label_clean.upper() and \
                   not any(pattern.search(cleaned_main_content_label) for pattern in compiled_sub_label_patterns) and \
                   not cleaned_main_content_label.upper().startswith('TELÉFONO'):
                    extracted_person_name_candidates.append(re.sub(r':$', '', cleaned_main_content_label).strip())


            # Iterate through all other columns in the same row
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            extracted_person_phone_candidates.append(phone_match_in_cell.group(1).strip())
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and name_part.upper() != base_field_label_clean.upper():
                                extracted_person_name_candidates.append(name_part)
                        else:
                            # If no phone, and it's not a sub-label pattern, and not the base label itself, it could be a name
                            if cleaned_item_content_val.upper() != base_field_label_clean.upper() and \
                               not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO'):
                                extracted_person_name_candidates.append(cleaned_item_content_val)

            # Assign the best candidates. Prioritize non-generic names.
            person_name = None
            for name_cand in extracted_person_name_candidates:
                if name_cand and name_cand.upper() != base_field_label_clean.upper(): # Only consider if it's not the generic label
                    person_name = name_cand
                    break # Take the first actual name found

            person_phone = next((s for s in extracted_person_phone_candidates if s), None) # Take first phone found

            el_rosario_extracted_data[f'{base_field_label_clean} Nombre'] = person_name
            el_rosario_extracted_data[f'{base_field_label_clean} Teléfono'] = person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # No `is_numerical_label = False` based on age_range_pattern in the label

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

````python
import re
import pandas as pd
import os
from google.colab import drive

# 1. Montar Google Drive (from cell IMFd_DZKWtGb)
print("Conectando con Google Drive...")
drive.mount('/content/drive')

# 2. Buscar la ruta del archivo automáticamente
ruta_archivo = None
for root, dirs, files in os.walk('/content/drive/My Drive'):
    if 'IPE_CARACTERIZACION.csv' in files:
        ruta_archivo = os.path.join(root, 'IPE_CARACTERIZACION.csv')
        break

if ruta_archivo:
    print(f"✅ Archivo encontrado en: {ruta_archivo}")

    # 3. Cargar el archivo con la lógica de limpieza
    try:
        df_carac = pd.read_csv(ruta_archivo, encoding='latin-1', sep=None, engine='python')

        for i in range(min(len(df_carac), 20)):
            if 'MUNICIPIO' in str(df_carac.iloc[i].values).upper():
                df_carac.columns = df_carac.iloc[i]
                df_carac = df_carac.iloc[i+1:].reset_index(drop=True)
                break

        df_carac.columns = [str(c).strip().upper() for c in df_carac.columns]
        print(f"📊 Se cargaron {len(df_carac)} registros de caracterización.")

    except Exception as e:
        print(f"❌ Error al leer el archivo: {e}")
el_carac = df_carac # Renaming to ensure consistency if variable was changed upstream

# Re-segmentation logic to define segmented_planillas (from cells f25e15a0 and 58678eab)
df_carac_col_upper = df_carac['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].str.upper().fillna('')

start_indices = df_carac_col_upper[df_carac_col_upper.str.contains('CARACTERIZACI\u00c3\u0093N.*CONSEJO COMUNAL', regex=True)].index.tolist()

end_indices = [idx - 1 for idx in start_indices[1:]] + [len(df_carac) - 1]

segmented_planillas = []
for i in range(len(start_indices)): # Use len(start_indices) as loop limit
    start = start_indices[i]
    end = end_indices[i]
    segmented_planillas.append(df_carac.iloc[start : end + 1].copy())


el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u0099', "'").replace('\u00e2\u0080\u0094', '-').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# --- Logic to find el_rosario_planilla (missing in previous execution context) ---
el_rosario_planilla = None
cc_name_pattern_search = re.compile(r'CARACTERIZACI\u00c3\u0093N DEL CONSEJO COMUNAL\s*(.*)', re.IGNORECASE)

for planilla in segmented_planillas:
    if 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' in planilla.columns:
        first_cell_content_raw = planilla['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].iloc[0]
        first_cell_content = clean_cell_content(first_cell_content_raw)
        if first_cell_content:
            match = cc_name_pattern_search.match(first_cell_content.upper())
            if match:
                name_raw = match.group(1).strip()
                # Re-clean name_raw to handle any encoding issues that might appear in the extracted group
                name_cleaned = clean_cell_content(name_raw)
                if name_cleaned == 'EL ROSARIO' or name_cleaned == 'EL ROSARIO"': # Handle slight variation in name
                    el_rosario_planilla = planilla
                    break
# ----------------------------------------------------------------------------------

if el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f"- {base_label.replace('N°', 'NN°')}") # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as simple numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_field_label_clean = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            person_name = None
            person_phone = None

            # Check if the main label itself contains name/phone (less common, but possible)
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                person_phone = phone_match_in_label.group(1).strip()
                name_candidate_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_candidate_from_label and name_candidate_from_label.upper() != base_field_label_clean.upper():
                    person_name = name_candidate_from_label

            # Iterate through all other columns in the same row to find the name and phone
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            if person_phone is None: # Assign phone if not already found
                                person_phone = phone_match_in_cell.group(1).strip()
                            # Try to extract name from the same cell if a name hasn't been found yet
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and person_name is None and not any(pattern.search(name_part) for pattern in compiled_sub_label_patterns):
                                person_name = name_part
                            # If both name and phone are found, we can break early
                            if person_name is not None and person_phone is not None:
                                break # Found both, exit column iteration for this row

                        elif person_name is None: # If no phone in this cell, and name is still missing
                            # Ensure it's not another label/sub-question or "TELÉFONO:" and not the base label itself
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO') and \
                               cleaned_item_content_val.upper() != base_field_label_clean.upper():
                                person_name = cleaned_item_content_val
                                # If we found a name, break
                                if person_phone is not None: # If phone was already found, we can break
                                    break

            # Assign extracted values to the final dictionary
            el_rosario_extracted_data[f'{base_field_label_clean} Nombre'] = person_name
            el_rosario_extracted_data[f'{base_field_label_clean} Teléfono'] = person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # No `is_numerical_label = False` based on age_range_pattern in the label

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

**Reasoning**:
The previous code failed because the regex pattern used to find 'el_rosario_planilla' was expecting the original encoded characters, but the string it was matching against had already been cleaned. This caused the pattern to not match any planilla. I will correct the `cc_name_pattern_search` regex to use the cleaned version of 'CARACTERIZACIÓN' so that it correctly matches the cleaned `first_cell_content`. Additionally, I will add `“` and `”` to the `clean_cell_content` function for better handling of quotation marks, as observed in previous outputs.



````python
import re
import pandas as pd
import os
from google.colab import drive

# 1. Montar Google Drive (from cell IMFd_DZKWtGb)
print("Conectando con Google Drive...")
drive.mount('/content/drive')

# 2. Buscar la ruta del archivo automáticamente
ruta_archivo = None
for root, dirs, files in os.walk('/content/drive/My Drive'):
    if 'IPE_CARACTERIZACION.csv' in files:
        ruta_archivo = os.path.join(root, 'IPE_CARACTERIZACION.csv')
        break

if ruta_archivo:
    print(f"✅ Archivo encontrado en: {ruta_archivo}")

    # 3. Cargar el archivo con la lógica de limpieza
    try:
        df_carac = pd.read_csv(ruta_archivo, encoding='latin-1', sep=None, engine='python')

        for i in range(min(len(df_carac), 20)):
            if 'MUNICIPIO' in str(df_carac.iloc[i].values).upper():
                df_carac.columns = df_carac.iloc[i]
                df_carac = df_carac.iloc[i+1:].reset_index(drop=True)
                break

        df_carac.columns = [str(c).strip().upper() for c in df_carac.columns]
        print(f"📊 Se cargaron {len(df_carac)} registros de caracterización.")

    except Exception as e:
        print(f"❌ Error al leer el archivo: {e}")
el_carac = df_carac # Renaming to ensure consistency if variable was changed upstream

# Re-segmentation logic to define segmented_planillas (from cells f25e15a0 and 58678eab)
df_carac_col_upper = df_carac['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].str.upper().fillna('')

start_indices = df_carac_col_upper[df_carac_col_upper.str.contains('CARACTERIZACI\u00c3\u0093N.*CONSEJO COMUNAL', regex=True)].index.tolist()

end_indices = [idx - 1 for idx in start_indices[1:]] + [len(df_carac) - 1]

segmented_planillas = []
for i in range(len(start_indices)): # Use len(start_indices) as loop limit
    start = start_indices[i]
    end = end_indices[i]
    segmented_planillas.append(df_carac.iloc[start : end + 1].copy())


el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u0099', "'").replace('\u00e2\u0080\u0094', '-').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"').replace('\u201c', '"').replace('\u201d', '"')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# --- Logic to find el_rosario_planilla (missing in previous execution context) ---
el_rosario_planilla = None
# CORRECTED: This regex pattern now expects the CLEANED version of 'CARACTERIZACIÓN'
cc_name_pattern_search = re.compile(r'CARACTERIZACIÓN DEL CONSEJO COMUNAL\s*(.*)', re.IGNORECASE)

for i, planilla in enumerate(segmented_planillas):
    if 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' in planilla.columns:
        first_cell_content_raw = planilla['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].iloc[0]
        first_cell_content_cleaned = clean_cell_content(first_cell_content_raw)

        if first_cell_content_cleaned:
            # Match against the cleaned content
            match = cc_name_pattern_search.match(first_cell_content_cleaned.upper())
            if match:
                name_raw = match.group(1).strip()
                name_cleaned = clean_cell_content(name_raw)

                # Debug print to see what is being compared
                # print(f"Planilla {i}: Found base name '{name_cleaned}' for comparison.")

                # The actual name of the CC is often in the row AFTER the 'CARACTERIZACIÓN DEL CONSEJO COMUNAL' row.
                # However, the current logic is relying on `match.group(1)` which directly extracts the name
                # if it's appended to the 'CARACTERIZACIÓN DEL CONSEJO COMUNAL' string itself.
                # Let's ensure the match is exact for 'EL ROSARIO'.
                # The name in df_viviendas.unique() is 'EL ROSARIO'.

                # To make sure we correctly identify the 'EL ROSARIO' planilla,
                # we check if the extracted name (which is part of the first cell content in the planilla) is 'EL ROSARIO'
                if name_cleaned == 'EL ROSARIO':
                    el_rosario_planilla = planilla
                    # print(f"Found 'EL ROSARIO' at planilla {i}")
                    break
# ----------------------------------------------------------------------------------

if el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. Please ensure the previous step to find the planilla for 'EL ROSARIO' was executed successfully.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f"- {base_label.replace('N°', 'NN°')}") # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_field_label_clean = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            person_name = None
            person_phone = None

            # Check if the main label itself contains name/phone (less common, but possible)
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                person_phone = phone_match_in_label.group(1).strip()
                name_candidate_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_candidate_from_label and name_candidate_from_label.upper() != base_field_label_clean.upper():
                    person_name = name_candidate_from_label

            # Iterate through all other columns in the same row to find the name and phone
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            if person_phone is None: # Assign phone if not already found
                                person_phone = phone_match_in_cell.group(1).strip()
                            # Try to extract name from the same cell if a name hasn't been found yet
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and person_name is None and not any(pattern.search(name_part) for pattern in compiled_sub_label_patterns):
                                person_name = name_part
                            # If both name and phone are found, we can break early
                            if person_name is not None and person_phone is not None:
                                break # Found both, exit column iteration for this row

                        elif person_name is None: # If no phone in this cell, and name is still missing
                            # Ensure it's not another label/sub-question or "TELÉFONO:" and not the base label itself
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO') and \
                               cleaned_item_content_val.upper() != base_field_label_clean.upper():
                                person_name = cleaned_item_content_val
                                # If we found a name, break
                                if person_phone is not None: # If phone was already found, we can break
                                    break

            # Assign extracted values to the final dictionary
            el_rosario_extracted_data[f'{base_field_label_clean} Nombre'] = person_name
            el_rosario_extracted_data[f'{base_field_label_clean} Teléfono'] = person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            # No `is_numerical_label = False` based on age_range_pattern in the label

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

Additionally, I will refine the `clean_cell_content` function to improve numerical parsing, and complete the logic for extracting contact names and phone numbers, ensuring all parts of the subtask are addressed for accurate data extraction.



````python
import re
import pandas as pd
import os
from google.colab import drive

# 1. Montar Google Drive (from cell IMFd_DZKWtGb)
print("Conectando con Google Drive...")
drive.mount('/content/drive')

# 2. Buscar la ruta del archivo automáticamente
ruta_archivo = None
for root, dirs, files in os.walk('/content/drive/My Drive'):
    if 'IPE_CARACTERIZACION.csv' in files:
        ruta_archivo = os.path.join(root, 'IPE_CARACTERIZACION.csv')
        break

if ruta_archivo:
    print(f"✅ Archivo encontrado en: {ruta_archivo}")

    # 3. Cargar el archivo con la lógica de limpieza
    try:
        df_carac = pd.read_csv(ruta_archivo, encoding='latin-1', sep=None, engine='python')

        for i in range(min(len(df_carac), 20)):
            if 'MUNICIPIO' in str(df_carac.iloc[i].values).upper():
                df_carac.columns = df_carac.iloc[i]
                df_carac = df_carac.iloc[i+1:].reset_index(drop=True)
                break

        df_carac.columns = [str(c).strip().upper() for c in df_carac.columns]
        print(f"📊 Se cargaron {len(df_carac)} registros de caracterización.")

    except Exception as e:
        print(f"❌ Error al leer el archivo: {e}")
else:
    print("❌ No pude encontrar el archivo 'IPE_CARACTERIZACION.csv' en tu Drive.")
    print("Asegúrate de que el nombre sea exacto o que esté en 'Mi Unidad'.")

# Re-segmentation logic to define segmented_planillas (from cells f25e15a0 and 58678eab)
df_carac_col_upper = df_carac['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].str.upper().fillna('')

start_indices = df_carac_col_upper[df_carac_col_upper.str.contains('CARACTERIZACI\u00c3\u0093N.*CONSEJO COMUNAL', regex=True)].index.tolist()

end_indices = [idx - 1 for idx in start_indices[1:]] + [len(df_carac) - 1]

segmented_planillas = []
for i in range(len(start_indices)): # Use len(start_indices) as loop limit
    start = start_indices[i]
    end = end_indices[i]
    segmented_planillas.append(df_carac.iloc[start : end + 1].copy())

el_rosario_extracted_data = {}
main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u0099', "'").replace('\u00e2\u0080\u0094', '-').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# --- Logic to find el_rosario_planilla (missing in previous execution context) ---
el_rosario_planilla = None
# Corrected: This regex pattern now expects the CLEANED version of 'CARACTERIZACIÓN'
cc_name_pattern_search = re.compile(r'CARACTERIZACIÓN DEL CONSEJO COMUNAL\s*(.*)', re.IGNORECASE)

for i, planilla in enumerate(segmented_planillas):
    if 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' in planilla.columns:
        first_cell_content_raw = planilla['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].iloc[0]
        first_cell_content_cleaned = clean_cell_content(first_cell_content_raw)

        if first_cell_content_cleaned:
            # Match against the cleaned content
            match = cc_name_pattern_search.match(first_cell_content_cleaned.upper())
            if match:
                name_raw = match.group(1).strip()
                name_cleaned = clean_cell_content(name_raw)

                # The actual name of the CC is often in the row AFTER the 'CARACTERIZACIÓN DEL CONSEJO COMUNAL' row.
                # However, the current logic is relying on `match.group(1)` which directly extracts the name
                # if it's appended to the 'CARACTERIZACIÓN DEL CONSEJO COMUNAL' string itself.
                # The name in df_viviendas.unique() was 'EL ROSARIO'. Let's match against the cleaned name.

                if name_cleaned == 'EL ROSARIO':
                    el_rosario_planilla = planilla
                    break

if el_rosario_planilla is None:
    print("Error: 'el_rosario_planilla' not found. This should not happen if previous steps were successful. Re-checking extraction logic.")
    # Fallback/Debug: Manually find if the name is in the second row, which is typical for 'Nombre completo del Consejo Comunal'
    for i, planilla in enumerate(segmented_planillas):
        if 'NOMBRE COMPLETO DEL CONSEJO COMUNAL' in planilla.columns and len(planilla) > 1:
            cc_name_from_second_row = clean_cell_content(planilla['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].iloc[1])
            if cc_name_from_second_row and ('EL ROSARIO' in cc_name_from_second_row.upper()):
                el_rosario_planilla = planilla
                print(f"Found EL ROSARIO using second row logic in planilla {i}")
                break

if el_rosario_planilla is None:
    print("Fatal Error: Could not find the planilla for 'EL ROSARIO' after all attempts.")
else:
    # Define labels that are expected to have numerical values
    numerical_value_labels = [
        'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
        'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
        'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
        'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
        'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
        'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
        'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
    ]
    # Pre-clean labels for consistent matching and convert to uppercase for comparison
    numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

    numerical_labels_to_match = set()
    for base_label in numerical_value_labels_base:
        numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
        numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
        numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
        numerical_labels_to_match.add(f"- {base_label.replace('N°', 'NN°')}") # e.g., - NN° DE FAMILIAS

    # Convert to tuple for efficient 'startswith' check
    numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

    # Define common header labels that often don't have direct values in the same row
    header_labels_cleaned = [clean_cell_content(label).upper() for label in [
        'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
        'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
        'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
        'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
    ]]

    # Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
    sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
    compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

    # Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
    compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

    # Pattern to identify age ranges, to avoid misinterpreting them as single numbers
    age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)

    processed_indices = set() # Keep track of indices that have been specifically handled

    # Iterate through the planilla
    for idx in range(len(el_rosario_planilla)):
        if idx in processed_indices:
            continue # Skip rows already handled by special logic

        row = el_rosario_planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in el_rosario_extracted_data:
                next_row_index = idx + 1
                if next_row_index < len(el_rosario_planilla):
                    next_row = el_rosario_planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    el_rosario_extracted_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            el_rosario_extracted_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            el_rosario_extracted_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in el_rosario_extracted_data:
                el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_field_label_clean = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            person_name = None
            person_phone = None

            # Check if the main label itself contains name/phone (less common, but possible)
            phone_match_in_label = re.search(r'(\d[\d\s\-\(\\ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                person_phone = phone_match_in_label.group(1).strip()
                name_candidate_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_candidate_from_label and name_candidate_from_label.upper() != base_field_label_clean.upper():
                    person_name = name_candidate_from_label

            # Iterate through all other columns in the same row to find the name and phone
            for col_name_val in el_rosario_planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\(\\ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            if person_phone is None: # Assign phone if not already found
                                person_phone = phone_match_in_cell.group(1).strip()
                            # Try to extract name from the same cell if a name hasn't been found yet
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and person_name is None and not any(pattern.search(name_part) for pattern in compiled_sub_label_patterns):
                                person_name = name_part
                            # If both name and phone are found, we can break early
                            if person_name is not None and person_phone is not None:
                                break # Found both, exit column iteration for this row

                        elif person_name is None: # If no phone in this cell, and name is still missing
                            # Ensure it's not another label/sub-question or "TELÉFONO:" and not the base label itself
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO') and \
                               cleaned_item_content_val.upper() != base_field_label_clean.upper():
                                person_name = cleaned_item_content_val
                                # If we found a name, break
                                if person_phone is not None: # If phone was already found, we can break
                                    break

            # Assign extracted values to the final dictionary
            el_rosario_extracted_data[f'{base_field_label_clean} Nombre'] = person_name
            el_rosario_extracted_data[f'{base_field_label_clean} Teléfono'] = person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in el_rosario_extracted_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in el_rosario_planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            el_rosario_extracted_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in el_rosario_extracted_data:
                    el_rosario_extracted_data[cleaned_main_content_label] = None
            processed_indices.add(idx) # Mark this row as processed by general logic

    # Print the extracted data in a readable format
    print("Información detallada para el Consejo Comunal 'EL ROSARIO' (Final Refinamiento):")
    print("-----------------------------------------------------------------------")
    for field, value in el_rosario_extracted_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")
````

### Subtask:
Iterate through each `planilla` in the `segmented_planillas` list. The subtask requires iterating through all `planilla` dataframes in `segmented_planillas` and applying the refined extraction logic. This will store the extracted data for each communal council in a list of dictionaries.



````python
import re
import pandas as pd

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u0099', "'").replace('\u00e2\u0080\u0094', '-').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^\s*(\d{1,3}(?:\.\d{3})*)(?:\s*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?\s*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Define labels that are expected to have numerical values
numerical_value_labels = [
    'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
    'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
    'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
    'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
    'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
    'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
    'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
]
# Pre-clean labels for consistent matching and convert to uppercase for comparison
numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

numerical_labels_to_match = set()
for base_label in numerical_value_labels_base:
    numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
    numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
    numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
    numerical_labels_to_match.add(f"- {base_label.replace('N°', 'NN°')}") # e.g., - NN° DE FAMILIAS

# Convert to tuple for efficient 'startswith' check
numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

# Define common header labels that often don't have direct values in the same row
header_labels_cleaned = [clean_cell_content(label).upper() for label in [
    'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
    'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
    'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
    'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
    'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
    'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
]]

# Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

# Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

# Pattern to identify age ranges, to avoid misinterpreting them as single numbers
age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)


all_extracted_data = [] # List to store extracted data for all planillas

# Iterate through each planilla in the segmented_planillas list
for planilla_index, planilla in enumerate(segmented_planillas):
    extracted_council_data = {}
    processed_indices = set()
    main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

    if main_content_col_name not in planilla.columns:
        # Skip this planilla if the main content column is missing
        print(f"Skipping planilla {planilla_index} due to missing '{main_content_col_name}' column.")
        continue

    for idx in range(len(planilla)):
        if idx in processed_indices:
            continue

        row = planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in extracted_council_data:
                extracted_council_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in extracted_council_data:
                next_row_index = idx + 1
                if next_row_index < len(planilla):
                    next_row = planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    extracted_council_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            extracted_council_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            extracted_council_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in extracted_council_data:
                extracted_council_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_field_label_clean = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            person_name = None
            person_phone = None

            # Check if the main label itself contains name/phone (less common, but possible)
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                person_phone = phone_match_in_label.group(1).strip()
                name_candidate_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_candidate_from_label and name_candidate_from_label.upper() != base_field_label_clean.upper():
                    person_name = name_candidate_from_label

            # Iterate through all other columns in the same row to find the name and phone
            for col_name_val in planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            if person_phone is None: # Assign phone if not already found
                                person_phone = phone_match_in_cell.group(1).strip()
                            # Try to extract name from the same cell if a name hasn't been found yet
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and person_name is None and not any(pattern.search(name_part) for pattern in compiled_sub_label_patterns):
                                person_name = name_part
                            # If both name and phone are found, we can break early
                            if person_name is not None and person_phone is not None:
                                break # Found both, exit column iteration for this row

                        elif person_name is None: # If no phone in this cell, and name is still missing
                            # Ensure it's not another label/sub-question or "TELÉFONO:" and not the base label itself
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO') and \
                               cleaned_item_content_val.upper() != base_field_label_clean.upper():
                                person_name = cleaned_item_content_val
                                # If we found a name, break
                                if person_phone is not None:
                                    break

            # Assign extracted values to the final dictionary
            extracted_council_data[f'{base_field_label_clean} Nombre'] = person_name
            extracted_council_data[f'{base_field_label_clean} Teléfono'] = person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in extracted_council_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            extracted_council_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in extracted_council_data:
                    extracted_council_data[cleaned_main_content_label] = None
            processed_indices.add(idx)

    all_extracted_data.append(extracted_council_data)

print(f"Extracted data for {len(all_extracted_data)} communal councils.")

# Optional: Display a sample of the extracted data for the first few councils
for i, council_data in enumerate(all_extracted_data[:3]):
    print(f"\n--- Data for Council {i+1} ---")
    for field, value in council_data.items():
        if value is not None:
            print(f"- {field}: {value}")
        else:
            print(f"- {field}")

````

The previous code failed with a NameError because `df_consolidated` was not defined in the execution context. To fix this, I will regenerate the complete code block to ensure `df_carac` is loaded, `segmented_planillas` is created, the extraction logic is applied to generate `all_extracted_data`, and `df_consolidated` is properly assembled from `all_extracted_data` before proceeding with the final cleaning and standardization steps. This combined approach ensures all dependencies are met and the data is correctly processed.



````python
import re
import pandas as pd
import os
from google.colab import drive

# --- Start: Re-running previous steps to ensure df_consolidated is defined ---

# 1. Montar Google Drive (from cell IMFd_DZKWtGb)
print("Conectando con Google Drive...")
drive.mount('/content/drive')

# 2. Buscar la ruta del archivo automáticamente
ruta_archivo = None
for root, dirs, files in os.walk('/content/drive/My Drive'):
    if 'IPE_CARACTERIZACION.csv' in files:
        ruta_archivo = os.path.join(root, 'IPE_CARACTERIZACION.csv')
        break

if ruta_archivo:
    print(f"✅ Archivo encontrado en: {ruta_archivo}")

    # 3. Cargar el archivo con la lógica de limpieza
    try:
        df_carac = pd.read_csv(ruta_archivo, encoding='latin-1', sep=None, engine='python')

        for i in range(min(len(df_carac), 20)):
            if 'MUNICIPIO' in str(df_carac.iloc[i].values).upper():
                df_carac.columns = df_carac.iloc[i]
                df_carac = df_carac.iloc[i+1:].reset_index(drop=True)
                break

        df_carac.columns = [str(c).strip().upper() for c in df_carac.columns]
        print(f"📊 Se cargaron {len(df_carac)} registros de caracterización.")

    except Exception as e:
        print(f"❌ Error al leer el archivo: {e}")
else:
    print("❌ No pude encontrar el archivo 'IPE_CARACTERIZACION.csv' en tu Drive.")
    print("Asegúrate de que el nombre sea exacto o que esté en 'Mi Unidad'.")

# Re-segmentation logic to define segmented_planillas (from cells f25e15a0 and 58678eab)
df_carac_col_upper = df_carac['NOMBRE COMPLETO DEL CONSEJO COMUNAL'].str.upper().fillna('')

start_indices = df_carac_col_upper[df_carac_col_upper.str.contains('CARACTERIZACI\u00c3\u0093N.*CONSEJO COMUNAL', regex=True)].index.tolist()

end_indices = [idx - 1 for idx in start_indices[1:]] + [len(df_carac) - 1]

segmented_planillas = []
for i in range(len(start_indices)): # Use len(start_indices) as loop limit
    start = start_indices[i]
    end = end_indices[i]
    segmented_planillas.append(df_carac.iloc[start : end + 1].copy())

# Helper function to clean cell content, handling potential NaN values and Series inputs,
# and improved number extraction.
def clean_cell_content(content, expect_number=False):
    if isinstance(content, pd.Series):
        for item in content.values:
            cleaned_item = clean_cell_content(item, expect_number)
            if cleaned_item is not None:
                return cleaned_item
        return None

    if pd.isna(content):
        return None

    s_content = str(content).strip()
    if s_content.upper() in ['NAN', 'NONE', '']:
        return None

    s_content = s_content.replace('\u00c3\u0093', 'Ó').replace('\u00c3\u0081', 'Á').replace('\u00c3\u0089', 'É').replace('\u00c3\u00a1', 'á').replace('\u00c3\u00a9', 'é').replace('\u00c3\u00ad', 'í').replace('\u00c3\u00b3', 'ó').replace('\u00c3\u00ba', 'ú').replace('\u00c3\u00b1', 'ñ').replace('\u00c2\u00b0', 'N°').replace('\u00e2\u0080\u0099', "'").replace('\u00e2\u0080\u0094', '-').replace('\u00e2\u0080\u0093', '-').replace('\u00c2\u00bf', '¿').replace('\u00e2\u0080\u009c', '"').replace('\u00e2\u0080\u009d', '"')

    if expect_number:
        # Improved regex to capture full numbers including optional thousands separators ('.')
        # and optional descriptive text like 'habitantes', 'familias', etc., but only if it's a pure number or number+text combination.
        # This will NOT match ranges like '12-17'.
        number_pattern_strict = re.compile(r'^[\s]*(\d{1,3}(?:\.\d{3})*)(?:[\s]*(?:mil|millones|habitantes|familias|niños|viviendas|hectáreas|años|neonatos|lactantes|casas|edificios|quintas|módulos))?[\s]*$', re.IGNORECASE)
        match_number = number_pattern_strict.match(s_content) # Use .match() to ensure it matches from the beginning
        if match_number:
            try:
                # Remove thousands separators before converting to int
                num_str = re.sub(r'[^\d]', '', match_number.group(1))
                return int(num_str)
            except ValueError:
                pass
        return None # If strict number expected but not found or not parsable

    return s_content

# Define labels that are expected to have numerical values
numerical_value_labels = [
    'N° DE FAMILIAS', 'N° DE POBLACIÓN TOTAL', 'N° DE NEONATOS O LACTANTES (0-2 AÑOS)',
    'N° DE NIÑOS (3-11 AÑOS)', 'N° DE ADOLESCENTES (12-17 AÑOS)', 'N° DE ADULTOS (18-59 AÑOS)',
    'N° DE ADULTOS MAYORES (MÁS DE 60 AÑOS)', 'SUPERFICIE DEL CONSEJO COMUNAL (HECTÁREAS)',
    'N° TOTAL DE VIVIENDAS', 'N° DE CASAS TRADICIONALES', 'N° DE CASAS CONVENCIONALES',
    'N° DE QUINTAS', 'N° DE EDIFICIOS', 'N° DE EDIFICIOS DE LA GMVV',
    'N° DE CASAS DE LA GMVV', 'N° DE VIVIENDAS CON MATERIALES INADECUADOS (RANCHOS)',
    'N° DE OTROS TIPOS DE VIVIENDAS (INDICAR SU TIPO)', 'N° DE VIVIENDAS EN RIESGO (ESPECIFICAR QUÉ TIPO DE VIVIENDA)'
]
# Pre-clean labels for consistent matching and convert to uppercase for comparison
numerical_value_labels_base = [clean_cell_content(label).upper() for label in numerical_value_labels]

numerical_labels_to_match = set()
for base_label in numerical_value_labels_base:
    numerical_labels_to_match.add(base_label) # e.g., N° DE FAMILIAS
    numerical_labels_to_match.add(base_label.replace('N°', 'NN°')) # e.g., NN° DE FAMILIAS
    numerical_labels_to_match.add(f'- {base_label}') # e.g., - N° DE FAMILIAS
    numerical_labels_to_match.add(f"- {base_label.replace('N°', 'NN°')}") # e.g., - NN° DE FAMILIAS

# Convert to tuple for efficient 'startswith' check
numerical_labels_to_match_tuple = tuple(numerical_labels_to_match)

# Define common header labels that often don't have direct values in the same row
header_labels_cleaned = [clean_cell_content(label).upper() for label in [
    'INFORMACIÓN ORGANIZACIONAL', 'INFORMACIÓN DEMOGRÁFICA',
    'INFORMACIÓN ECONÓMICA', 'INFORMACIÓN SOCIAL',
    'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ABRIL 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
    'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE AGOSTO 2024 PROYECTOS PRIORIZADOS EN LOS QUE HA PARTICIPADO',
    'INFORMACIÓN SOBRE CONSULTA POPULAR NACIONAL DE ENERO 2025 PROYECTOS PRIORIZADOS EN LOS QUE PARTICIPO EL CONSEJO COMUNAL',
    'INSTITUTO DE PROYECTOS ESPECIALES (IPE)'
]]

# Patterns to identify if content is a sub-label or question (e.g., starts with '-', or contains 'N°', or ends with '?')
sub_label_patterns = [r'^-', r'\?$', r'INDICAR SU TIPO', r'1ER', r'2DO', r'3ER', r'4TO', r'5TO', r'6TO', r'7MO']
compiled_sub_label_patterns = [re.compile(p, re.IGNORECASE) for p in sub_label_patterns]

# Added 'TELÉFONO:' to the list of patterns that identify a label (not a value)
compiled_sub_label_patterns.append(re.compile(r'TELÉFONO:', re.IGNORECASE))

# Pattern to identify age ranges, to avoid misinterpreting them as single numbers
age_range_pattern = re.compile(r'\(\d+\s*-\s*\d+\s*(?:años)?\)', re.IGNORECASE)


all_extracted_data = [] # List to store extracted data for all planillas

# Iterate through each planilla in the segmented_planillas list
for planilla_index, planilla in enumerate(segmented_planillas):
    extracted_council_data = {}
    processed_indices = set()
    main_content_col_name = 'NOMBRE COMPLETO DEL CONSEJO COMUNAL'

    if main_content_col_name not in planilla.columns:
        # Skip this planilla if the main content column is missing
        print(f"Skipping planilla {planilla_index} due to missing '{main_content_col_name}' column.")
        continue

    for idx in range(len(planilla)):
        if idx in processed_indices:
            continue

        row = planilla.iloc[idx]
        cleaned_main_content_label = clean_cell_content(row[main_content_col_name])

        if cleaned_main_content_label is None:
            continue

        # Special case: The very first header identifying the communal council
        if cleaned_main_content_label.upper().startswith("CARACTERIZACIÓN DEL CONSEJO COMUNAL"):
            if cleaned_main_content_label not in extracted_council_data:
                extracted_council_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Special handling for 'Nombre completo del Consejo Comunal', 'Parroquia y Municipio', 'Estado'
        if cleaned_main_content_label == 'Nombre completo del Consejo Comunal':
            if cleaned_main_content_label not in extracted_council_data:
                next_row_index = idx + 1
                if next_row_index < len(planilla):
                    next_row = planilla.iloc[next_row_index]

                    cc_name_value = clean_cell_content(next_row[main_content_col_name])
                    extracted_council_data[cleaned_main_content_label] = cc_name_value

                    if 'PARROQUIA Y MUNICIPIO' in next_row.index:
                        p_y_m_value = clean_cell_content(next_row['PARROQUIA Y MUNICIPIO'])
                        if p_y_m_value is not None:
                            extracted_council_data['Parroquia y Municipio'] = p_y_m_value

                    if 'ESTADO' in next_row.index:
                        estado_value = clean_cell_content(next_row['ESTADO'])
                        if estado_value is not None:
                            extracted_council_data['Estado'] = estado_value

                processed_indices.add(idx)
                processed_indices.add(next_row_index)
            continue

        # If the main content is a recognized header, simply add it as a key with None value if not already present
        if cleaned_main_content_label.upper() in header_labels_cleaned:
            if cleaned_main_content_label not in extracted_council_data:
                extracted_council_data[cleaned_main_content_label] = None
            processed_indices.add(idx)
            continue

        # Handle contact information fields separately
        if cleaned_main_content_label.upper().startswith('REPRESENTANTE COMUNAL') or \
           cleaned_main_content_label.upper().startswith('RESPONSABLE DEL REGISTRO'):

            base_field_label_clean = re.sub(r':\s*TELÉFONO:$', '', cleaned_main_content_label, flags=re.IGNORECASE).strip()

            person_name = None
            person_phone = None

            # Check if the main label itself contains name/phone (less common, but possible)
            phone_match_in_label = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_main_content_label, re.IGNORECASE)
            if phone_match_in_label:
                person_phone = phone_match_in_label.group(1).strip()
                name_candidate_from_label = cleaned_main_content_label.replace(phone_match_in_label.group(0), '', 1).replace(':', '').strip()
                if name_candidate_from_label and name_candidate_from_label.upper() != base_field_label_clean.upper():
                    person_name = name_candidate_from_label

            # Iterate through all other columns in the same row to find the name and phone
            for col_name_val in planilla.columns:
                if col_name_val != main_content_col_name:
                    item_content_val = row[col_name_val]
                    cleaned_item_content_val = clean_cell_content(item_content_val) # Get the raw cleaned string

                    if cleaned_item_content_val is not None:
                        phone_match_in_cell = re.search(r'(\d[\d\s\-\( ext.]+)', cleaned_item_content_val, re.IGNORECASE)
                        if phone_match_in_cell:
                            if person_phone is None: # Assign phone if not already found
                                person_phone = phone_match_in_cell.group(1).strip()
                            # Try to extract name from the same cell if a name hasn't been found yet
                            name_part = cleaned_item_content_val.replace(phone_match_in_cell.group(0), '', 1).strip()
                            if name_part and person_name is None and not any(pattern.search(name_part) for pattern in compiled_sub_label_patterns):
                                person_name = name_part
                            # If both name and phone are found, we can break early
                            if person_name is not None and person_phone is not None:
                                break # Found both, exit column iteration for this row

                        elif person_name is None: # If no phone in this cell, and name is still missing
                            # Ensure it's not another label/sub-question or "TELÉFONO:" and not the base label itself
                            if not any(pattern.search(cleaned_item_content_val) for pattern in compiled_sub_label_patterns) and \
                               not cleaned_item_content_val.upper().startswith('TELÉFONO') and \
                               cleaned_item_content_val.upper() != base_field_label_clean.upper():
                                person_name = cleaned_item_content_val
                                # If we found a name, break
                                if person_phone is not None:
                                    break

            # Assign extracted values to the final dictionary
            extracted_council_data[f'{base_field_label_clean} Nombre'] = person_name
            extracted_council_data[f'{base_field_label_clean} Teléfono'] = person_phone
            processed_indices.add(idx)
            continue

        # General logic for all other fields: search for value in other columns of the current row
        if cleaned_main_content_label not in extracted_council_data:
            value_found_in_current_row = False

            # Check if the cleaned_main_content_label starts with any of the known numerical patterns
            is_numerical_label = any(cleaned_main_content_label.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

            is_main_label_a_sub_question = False
            for pattern in compiled_sub_label_patterns:
                if pattern.search(cleaned_main_content_label):
                    is_main_label_a_sub_question = True
                    break

            for col_name in planilla.columns:
                if col_name != main_content_col_name:
                    item_content = row[col_name]
                    # First, clean the item content without assuming it's a number
                    cleaned_item_content_raw = clean_cell_content(item_content)

                    if cleaned_item_content_raw is not None:
                        # Before trying to extract as a number, check if the raw cleaned content is actually another label
                        is_item_content_a_sub_question = False
                        for pattern in compiled_sub_label_patterns:
                            if pattern.search(cleaned_item_content_raw):
                                is_item_content_a_sub_question = True
                                break

                        # Also check if the raw content is a header or numerical label itself
                        is_item_content_a_header_or_num_label = cleaned_item_content_raw.upper() in header_labels_cleaned or \
                                                                any(cleaned_item_content_raw.upper().startswith(num_label_prefix) for num_label_prefix in numerical_labels_to_match_tuple)

                        # If the potential value is itself another label, skip it.
                        if (is_main_label_a_sub_question and is_item_content_a_sub_question) or \
                           is_item_content_a_header_or_num_label:
                            continue # Don't assign this as a value, continue searching for a proper value

                        # Now, if it's not a label, try to extract its value, potentially as a number.
                        extracted_value = clean_cell_content(item_content, expect_number=is_numerical_label)

                        if extracted_value is not None:
                            extracted_council_data[cleaned_main_content_label] = extracted_value
                            value_found_in_current_row = True
                            break # Found a valid value, stop searching for this label

            if not value_found_in_current_row:
                if cleaned_main_content_label not in extracted_council_data:
                    extracted_council_data[cleaned_main_content_label] = None
            processed_indices.add(idx)

    all_extracted_data.append(extracted_council_data)

# Create the df_consolidated DataFrame (from cell d3cb7d1e)
df_consolidated = pd.DataFrame(all_extracted_data)

print("Consolidated DataFrame created successfully before final cleaning.")

# --- End: Re-running previous steps ---

numeric_columns = [
    '- NN° de familias',
    '- NN° de población total',
    '- NN° de neonatos o lactantes (0-2 años)',
    '- NN° de niños (3-11 años)',
    '- NN° de adolescentes (12-17 años)',
    '- NN° de adultos (18-59 años)',
    '- NN° de adultos mayores (más de 60 años)',
    '- Superficie del consejo comunal (hectáreas)',
    '- NN° total de viviendas',
    '- NN° de casas tradicionales::',
    '- NN° de casas convencionales',
    '- NN° de quintas',
    '- NN° de edificios',
    '- NN° de edificios de la GMVV',
    '- NN° de casas de la GMVV',
    '- NN° de viviendas con materiales inadecuados\n(ranchos)',
    '- NN° de otros tipos de viviendas (indicar su tipo)',
    '- NN° de viviendas en riesgo (especificar qué tipo de vivienda)'
]

for col in numeric_columns:
    if col in df_consolidated.columns:
        df_consolidated[col] = pd.to_numeric(df_consolidated[col], errors='coerce')
    else:
        print(f"Warning: Column '{col}' not found in df_consolidated. Skipping conversion.")

# Fill any remaining NaN values in the entire DataFrame with 0
df_consolidated = df_consolidated.fillna(0)

print("Final cleaning and standardization complete.")
display(df_consolidated.head())
print("\nDataFrame Info after cleaning:")
df_consolidated.info()
````
